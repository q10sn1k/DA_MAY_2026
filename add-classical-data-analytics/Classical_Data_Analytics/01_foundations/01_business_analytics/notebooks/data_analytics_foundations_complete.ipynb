# 1. Понимание структуры данных: список

В аналитике данные часто приходят не по одному значению, а набором. Например:

- список сумм заказов;
- список возрастов клиентов;
- список оценок удовлетворенности;
- список товаров в заказе.

В Python для таких наборов часто используется **список** — `list`.

Индексация в Python начинается с нуля:

| Позиция для человека | Индекс в Python |
|---:|---:|
| первый элемент | `0` |
| второй элемент | `1` |
| третий элемент | `2` |

````python
# Создаем список значений
numbers = [12, 34, 56, 78]

# Выводим второй элемент списка
second_element = numbers[1]
print("Второй элемент списка:", second_element)

# Изменяем третий элемент списка на 100
numbers[2] = 100

# Выводим измененный список
print("Измененный список:", numbers)
````

### Разбор

- `numbers = [12, 34, 56, 78]` создает список.
- `numbers[1]` обращается ко второму элементу.
- `numbers[2] = 100` заменяет третий элемент.

Для аналитика это похоже на исправление значения в небольшом наборе данных.

## Практика 1.1

Есть список сумм заказов. Нужно вывести первый заказ, последний заказ и заменить второй заказ на исправленное значение.

````python
orders = [1200, 850, 4300, 2100]

print("Первый заказ:", orders[0])
print("Последний заказ:", orders[-1])

orders[1] = 900
print("Список после исправления:", orders)
````

# 4. Условные операторы: четное или нечетное число

Условные операторы позволяют программе принимать решения.

В Python используется конструкция:

```python
if условие:
    действие_если_условие_истинно
else:
    действие_если_условие_ложно
```

Для проверки четности используется остаток от деления `%`.

````python
number = 17

if number % 2 == 0:
    print("Число четное")
else:
    print("Число нечетное")
````

### Где это может пригодиться аналитику

Условия используются не только для четности. Например, можно классифицировать заказ:

- если сумма заказа больше 5000 — крупный заказ;
- иначе — обычный заказ.

````python
order_sum = 7200

if order_sum >= 5000:
    order_type = "крупный заказ"
else:
    order_type = "обычный заказ"

print("Сумма заказа:", order_sum)
print("Тип заказа:", order_type)
````

# 9. Условные операторы в анализе данных

Теперь проверим, является ли число положительным, отрицательным или нулем.

````python
value = -15

if value > 0:
    result = "положительное число"
elif value < 0:
    result = "отрицательное число"
else:
    result = "ноль"

print("Значение:", value)
print("Результат:", result)
````

## Бизнес-пример: отклонение от плана

Если факт больше плана — перевыполнение. Если меньше — отставание. Если равен — план выполнен точно.

````python
plan = 100000
fact = 93000

difference = fact - plan

if difference > 0:
    status = "план перевыполнен"
elif difference < 0:
    status = "есть отставание от плана"
else:
    status = "план выполнен точно"

print("План:", plan)
print("Факт:", fact)
print("Отклонение:", difference)
print("Статус:", status)
````

# 10. Создание и использование функций

Функция — это именованный блок кода, который можно вызывать много раз.

Функции полезны, когда один и тот же расчет повторяется:

- расчет площади;
- расчет скидки;
- расчет среднего чека;
- классификация клиента;
- проверка качества данных.

````python
import math

def calculate_area(radius):
    area = math.pi * radius ** 2
    return area

radius = 5
circle_area = calculate_area(radius)

print(f"Радиус: {radius}")
print(f"Площадь круга: {circle_area:.2f}")
````

### Важно: `return` не просто печатает значение

- `print()` показывает значение на экране.
- `return` возвращает значение из функции, чтобы его можно было использовать дальше.

````python
def calculate_revenue(price, quantity):
    return price * quantity

revenue = calculate_revenue(1500, 4)
tax = revenue * 0.2

print("Выручка:", revenue)
print("Налог 20%:", tax)
````

# 12. Цикл `for`: повторение действий

# 14. Цикл `while`

`while` выполняется, пока условие истинно.

В исходном задании нужно посчитать от 10 до 1 и вывести `Старт!`.

````python
counter = 10

while counter >= 1:
    print(counter)
    counter -= 1

print("Старт!")
````

## Важно

У цикла `while` обязательно должно быть изменение условия. Если забыть `counter -= 1`, цикл может стать бесконечным.

# 15. Преобразование цикла `for` в `while`

Цикл:

```python
for i in range(5):
    print(i)
```

можно переписать через `while`.

````python
i = 0

while i < 5:
    print(i)
    i += 1
````

# 16. Вложенные циклы

Вложенный цикл — это цикл внутри другого цикла.

Такой подход используется, когда нужно перебрать комбинации значений.

````python
for outer_number in range(1, 4):
    for inner_number in range(1, 3):
        print(f"Внешнее число: {outer_number}, внутреннее число: {inner_number}")
````

## Бизнес-пример: комбинации городов и категорий

````python
cities = ["Москва", "Казань", "Екатеринбург"]
categories = ["Электроника", "Одежда"]

for city in cities:
    for category in categories:
        print(f"Город: {city}, категория: {category}")
````

# 17. Работа со строками в pandas

Текстовые данные часто требуют предварительной обработки:

- привести имена к одному регистру;
- убрать лишние пробелы;
- проверить email;
- извлечь часть строки;
- привести числовые значения из текста к числам.

## 17.1. Приведение строк к нижнему регистру

Это помогает унифицировать данные. Например, `"ALICE"`, `"Alice"` и `"alice"` можно привести к одному виду.

````python
df_names = pd.DataFrame({
    "Name": ["alice", "BOB", "Charlie"]
})

df_names["Name_lower"] = df_names["Name"].str.lower()

display(df_names)
````

## 17.2. Преобразование чисел в строки и обратно

Иногда данные приходят из файлов как текст, хотя по смыслу это числа.

````python
df_age = pd.DataFrame({
    "Age": [24, 31, 45]
})

df_age["Age_as_text"] = df_age["Age"].astype(str)
df_age["Age_as_number"] = pd.to_numeric(df_age["Age_as_text"])

display(df_age)
print(df_age.dtypes)
````

## 17.3. Регулярные выражения: фильтрация email

Регулярные выражения позволяют искать текст по шаблону.

В этом примере выберем email, которые заканчиваются на `.com`.

````python
df_email = pd.DataFrame({
    "Email": [
        "anna@example.com",
        "ivan@company.ru",
        "manager@test.com",
        "wrong_email",
        "support@service.org"
    ]
})

filtered_email = df_email[df_email["Email"].str.contains(r"\.com$", regex=True)]

display(filtered_email)
````

## 17.4. Извлечение данных из строки

Иногда нужные признаки «спрятаны» внутри одной текстовой строки.

````python
df_info = pd.DataFrame({
    "Info": [
        "Name: Anna, Age: 29",
        "Name: Ivan, Age: 35",
        "Name: Maria, Age: 41"
    ]
})

df_info[["Name", "Age"]] = df_info["Info"].str.extract(r"Name: (\w+), Age: (\d+)")
df_info["Age"] = pd.to_numeric(df_info["Age"])

display(df_info)
print(df_info.dtypes)
````

# 18. Series и DataFrame в pandas

`Series` — это один столбец данных.  
`DataFrame` — это таблица из нескольких столбцов.

## 18.1. Создание и базовая манипуляция с Series

````python
sales_series = pd.Series([10, 20, 30, 40, 50], name="sales")

# Добавим новое значение 60
sales_series = pd.concat([sales_series, pd.Series([60])], ignore_index=True)

display(sales_series)
````

## 18.2. Работа с DataFrame и добавление строки

В новых версиях pandas для добавления строки лучше использовать `pd.concat()`, а не устаревший подход через `.append()`.

````python
df_clients = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [24, 27, 22]
})

new_client = pd.DataFrame({
    "Name": ["Diana"],
    "Age": [30]
})

df_clients = pd.concat([df_clients, new_client], ignore_index=True)

display(df_clients)
````

# 22. Итоговый мини-проект: первичный анализ продаж

Теперь объединим изученные темы в маленький аналитический сценарий.

Ситуация: у нас есть данные о заказах. Нужно:

1. создать таблицу заказов;
2. посмотреть первые строки;
3. проверить типы данных;
4. рассчитать базовые показатели;
5. сгруппировать данные по городам и категориям;
6. построить график;
7. сформулировать вывод.

````python
orders_df = pd.DataFrame({
    "order_id": [101, 102, 103, 104, 105, 106, 107, 108],
    "date": [
        "2026-05-01", "2026-05-01", "2026-05-02", "2026-05-02",
        "2026-05-03", "2026-05-03", "2026-05-04", "2026-05-04"
    ],
    "city": ["Москва", "Казань", "Москва", "Екатеринбург", "Казань", "Москва", "Екатеринбург", "Казань"],
    "category": ["Электроника", "Одежда", "Одежда", "Электроника", "Дом", "Дом", "Одежда", "Электроника"],
    "revenue": [75000, 8500, 12000, 68000, 15000, 22000, 9300, 54000],
    "is_paid": [True, True, False, True, True, True, False, True]
})

orders_df["date"] = pd.to_datetime(orders_df["date"])

display(orders_df)
````

## 22.1. Первичный просмотр данных

````python
print("Первые строки:")
display(orders_df.head())

print("Типы данных:")
print(orders_df.dtypes)

print("Размер таблицы:")
print(orders_df.shape)
````

## 22.2. Базовые KPI

````python
total_revenue = orders_df["revenue"].sum()
orders_count = orders_df["order_id"].count()
average_order_value = orders_df["revenue"].mean()
paid_share = orders_df["is_paid"].mean()

print(f"Общая выручка: {total_revenue:,.0f}")
print(f"Количество заказов: {orders_count}")
print(f"Средний чек: {average_order_value:,.0f}")
print(f"Доля оплаченных заказов: {paid_share:.1%}")
````

## 22.3. Группировка по городам

````python
city_summary = (
    orders_df
    .groupby("city", as_index=False)
    .agg(
        orders_count=("order_id", "count"),
        total_revenue=("revenue", "sum"),
        average_order_value=("revenue", "mean")
    )
    .sort_values("total_revenue", ascending=False)
)

display(city_summary)
````

## 22.4. Сводная таблица по городам и категориям

Сводная таблица помогает быстро пересечь два измерения: например, город и категорию.

````python
pivot_revenue = pd.pivot_table(
    orders_df,
    values="revenue",
    index="city",
    columns="category",
    aggfunc="sum",
    fill_value=0
)

display(pivot_revenue)
````

## 22.5. Визуализация выручки по городам

````python
city_summary.plot(
    x="city",
    y="total_revenue",
    kind="bar",
    legend=False,
    title="Выручка по городам"
)

plt.xlabel("Город")
plt.ylabel("Выручка")
plt.xticks(rotation=0)
plt.grid(axis="y")
plt.show()
````

## 22.6. Автоматический простой вывод

Сформируем короткий аналитический вывод на основе данных.

````python
top_city_row = city_summary.iloc[0]
top_city = top_city_row["city"]
top_city_revenue = top_city_row["total_revenue"]

paid_orders = orders_df[orders_df["is_paid"] == True]
unpaid_orders = orders_df[orders_df["is_paid"] == False]

print(f"Больше всего выручки принес город: {top_city} ({top_city_revenue:,.0f}).")
print(f"Оплаченных заказов: {len(paid_orders)}.")
print(f"Неоплаченных заказов: {len(unpaid_orders)}.")

if len(unpaid_orders) > 0:
    print("Есть неоплаченные заказы — их стоит проверить отдельно.")
else:
    print("Все заказы оплачены.")
````
