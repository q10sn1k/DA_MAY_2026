## 4. Проверка текущей папки

Очень частая ошибка начинающих — файл данных есть на компьютере, но Python ищет его не в той папке.

Проверим текущую рабочую папку.

````python
from pathlib import Path

current_dir = Path.cwd()

print("Текущая рабочая папка:")
print(current_dir)
````

### Важное пояснение

Относительные пути считаются от текущей рабочей папки.

Например, путь:

```python
data/raw/sales.csv
```

означает:

> найди папку `data`, внутри нее папку `raw`, внутри нее файл `sales.csv`, начиная от текущей рабочей папки.

## 6. Проверка наличия файлов данных

Теперь проверим, что все учебные файлы лежат на месте.

````python
required_files = [
    data_dir / "sales.csv",
    data_dir / "products.xlsx",
    data_dir / "regions.json",
    data_dir / "clients.csv",
    data_dir / "web_table_sample.html",
]

print("\nПроверяем файлы:")

for file_path in required_files:
    if file_path.exists():
        print(f"OK:   {file_path}")
    else:
        print(f"FAIL: {file_path} — файл не найден")
````

### Как читать результат

- `OK` — файл найден.
- `FAIL` — файл не найден.

Если есть `FAIL`, проверьте:

1. Распакован ли архив с материалами.

## 10. Типовые ошибки

### Ошибка 1. `ModuleNotFoundError`

Пример:

```text
ModuleNotFoundError: No module named 'pandas'
```

Причина: библиотека не установлена.

Что делать:

```bash
pip install -r requirements.txt
```

---

### Ошибка 2. `FileNotFoundError`

Пример:

```text
FileNotFoundError: [Errno 2] No such file or directory: 'data/raw/sales.csv'
```

Причина: Python не видит файл по указанному пути.

Что проверить:

- текущую папку через `Path.cwd()`;
- наличие папки `data/raw`;
- правильность имени файла;
- не открыт ли ноутбук из другой папки.

---

### Ошибка 3. `UnicodeDecodeError`

Причина: файл сохранен в другой кодировке.

Что можно попробовать:

```python
pd.read_csv("file.csv", encoding="utf-8")
pd.read_csv("file.csv", encoding="cp1251")
```

---

### Ошибка 4. `ParserError`

Причина: CSV-файл имеет неожиданный разделитель или поврежденную структуру.

Что можно попробовать:

```python
pd.read_csv("file.csv", sep=";")
pd.read_csv("file.csv", sep=",")
```
