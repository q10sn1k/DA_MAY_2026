# Практическая работа: SQL-запросы и pandas DataFrame

Этот ноутбук закрепляет материал после файлов по MySQL и ноутбука **«SQL → pandas»**.

Главная идея практики:

> В SQL мы пишем запрос к базе данных, а в pandas выполняем похожие действия с таблицей, которая уже загружена в Python.

В этой работе используется учебная таблица `users`. Она похожа на таблицу, которую мы создавали в MySQL.


## Что нужно знать перед выполнением

Перед началом желательно понимать базовые SQL-команды:

| SQL-команда | Что делает | Похожее действие в pandas |
|---|---|---|
| `SELECT * FROM users` | показать всю таблицу | `users` |
| `SELECT name, age FROM users` | выбрать столбцы | `users[["name", "age"]]` |
| `WHERE age > 25` | отфильтровать строки | `users[users["age"] > 25]` |
| `ORDER BY balance DESC` | отсортировать | `users.sort_values("balance", ascending=False)` |
| `COUNT(*)` | посчитать строки | `len(users)` |
| `AVG(balance)` | среднее значение | `users["balance"].mean()` |
| `GROUP BY country` | группировка | `users.groupby("country")` |

Не нужно заучивать всё сразу. Важно видеть связь между SQL и Python.

# Дополнительный пример: как pandas может читать SQL-запрос

Этот блок можно показывать после выполнения основных заданий.

В реальной работе данные часто лежат в MySQL, PostgreSQL или другой базе данных. pandas умеет получать результат SQL-запроса и сразу превращать его в `DataFrame`.

Ниже используется встроенная учебная база SQLite, чтобы пример работал без отдельного сервера MySQL.


````python
import sqlite3

connection = sqlite3.connect(":memory:")
users.to_sql("users", connection, index=False, if_exists="replace")

query = """
SELECT name, age, country, balance
FROM users
WHERE balance >= 95000
ORDER BY balance DESC;
"""

sql_result = pd.read_sql_query(query, connection)
sql_result
````

Что важно понять:

- SQL-запрос выполняется внутри базы;
- результат запроса возвращается в Python;
- pandas превращает результат в `DataFrame`;
- дальше можно использовать обычные методы pandas.
