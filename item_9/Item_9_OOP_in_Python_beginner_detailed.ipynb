# 1. Что такое ООП

**ООП — объектно-ориентированное программирование**. Это способ организовать программу вокруг объектов.

Объект удобно представлять как сущность, у которой есть:

- **состояние** — данные объекта;
- **поведение** — действия, которые объект умеет выполнять.

Например, у машины есть состояние:

- марка;
- цвет;
- текущая скорость;
- количество топлива.

И поведение:

- завестись;
- ускориться;
- затормозить;
- показать информацию о себе.

В обычной программе можно хранить марку, цвет и скорость в отдельных переменных. Но когда машин становится много, отдельные переменные быстро превращаются в путаницу. Класс позволяет собрать связанные данные и действия в одном месте.

## 1.1. Зачем использовать ООП

ООП особенно полезно, когда в программе есть много однотипных сущностей:

- пользователи;
- товары;
- автомобили;
- банковские счета;
- документы;
- заказы;
- сотрудники;
- датасеты и отчёты.

Преимущества:

1. **Данные и действия находятся рядом.** Скорость машины и метод ускорения описаны в одном классе.
2. **Код легче переиспользовать.** Один класс `Car` позволяет создать тысячи машин.
3. **Код легче расширять.** Можно создать электромобиль на основе обычной машины.
4. **Правила легче контролировать.** Можно запретить отрицательную скорость или отрицательный баланс.
5. **Программа становится ближе к предметной области.** В коде появляются понятные сущности: `Car`, `Order`, `Customer`.

ООП не делает программу автоматически правильной. Оно лишь предоставляет инструменты для более удобной организации кода.

## 1.2. Четыре классических принципа ООП

| Принцип | Простое объяснение | Пример |
|---|---|---|
| Инкапсуляция | Объект сам контролирует свои данные | Машина не должна позволять установить скорость `-50` |
| Наследование | Новый класс получает возможности существующего класса | Электромобиль наследует свойства обычной машины |
| Полиморфизм | Один и тот же вызов работает по-разному для разных объектов | `move()` у машины едет, у лодки плывёт |
| Абстракция | Мы показываем главное и скрываем лишние детали | Водитель нажимает «старт», не управляя каждым поршнем двигателя |

В Python эти принципы реализуются не как жёсткие правила, а как набор языковых механизмов и соглашений.

# 2. Класс, объект и экземпляр

Это три основных слова, с которых начинается ООП.

## Класс

**Класс** — это описание будущих объектов. Его можно сравнить с чертежом машины.

Чертёж не является машиной. На нём только указано:

- какие данные будут у машины;
- какие действия она сможет выполнять.

## Объект

**Объект** — конкретная сущность, созданная по классу.

## Экземпляр

Слово **экземпляр** подчёркивает, к какому классу относится объект.

Фразы «объект класса `Car`» и «экземпляр класса `Car`» обычно означают одно и то же.

````python
class Car:  # Объявляем новый класс с именем Car.
    pass  # Пока класс ничего не хранит и ничего не умеет.

my_car = Car()  # Создаём первый объект класса Car и сохраняем его в переменную my_car.
your_car = Car()  # Создаём второй независимый объект того же класса Car.

print(type(my_car))  # Показываем тип первого объекта: он относится к классу Car.
print(isinstance(my_car, Car))  # Проверяем, является ли my_car экземпляром класса Car.
print(my_car is your_car)  # Проверяем, являются ли переменные ссылками на один объект.
````

### Что произошло в коде

Строка `class Car:` создала класс.

Строка `my_car = Car()` вызвала класс как функцию и создала объект.

Важно:

- `Car` — класс;
- `Car()` — создание нового объекта;
- `my_car` — переменная, которая хранит ссылку на объект;
- `your_car` — другая переменная с другим объектом.

Даже если два объекта созданы одним классом, они остаются разными объектами.

## 2.1. Атрибуты — данные объекта

**Атрибут** — значение, связанное с объектом.

Для машины атрибутами могут быть:

- `brand` — марка;
- `color` — цвет;
- `speed` — скорость.

К атрибуту обращаются через точку:

```python
my_car.color
```

Точка читается как «у объекта `my_car` взять атрибут `color`».

````python
my_car.brand = "Toyota"  # Добавляем объекту my_car атрибут brand со значением Toyota.
my_car.color = "красный"  # Добавляем объекту my_car атрибут color со значением красный.
my_car.speed = 0  # Добавляем объекту my_car атрибут speed с начальной скоростью ноль.

your_car.brand = "BMW"  # Добавляем другому объекту собственную марку BMW.
your_car.color = "синий"  # Добавляем другому объекту собственный цвет.
your_car.speed = 20  # Добавляем другому объекту собственную скорость.

print(my_car.brand, my_car.color, my_car.speed)  # Выводим состояние первой машины.
print(your_car.brand, your_car.color, your_car.speed)  # Выводим состояние второй машины.
````

Добавлять атрибуты после создания объекта возможно, но это неудобно:

- можно забыть добавить атрибут;
- можно ошибиться в названии;
- разные объекты могут получить разную структуру.

Поэтому начальные атрибуты обычно создают внутри специального метода `__init__()`.

# 3. `self` и метод `__init__()`

## Что такое `__init__()`

`__init__()` — специальный метод, который автоматически вызывается после создания нового объекта.

Он используется для начальной настройки объекта.

Например, при создании машины мы сразу передаём:

- марку;
- цвет.

А скорость устанавливаем в ноль.

## Что такое `self`

`self` — ссылка на конкретный объект, с которым сейчас работает метод.

Когда создаются две машины, один и тот же метод `__init__()` вызывается два раза:

- первый раз `self` указывает на первую машину;
- второй раз `self` указывает на вторую машину.

Имя `self` является соглашением Python. Теоретически можно использовать другое имя, но так делать не принято.

````python
class Car:  # Повторно объявляем более полезную версию класса Car.
    def __init__(self, brand: str, color: str) -> None:  # Описываем начальную настройку новой машины.
        self.brand = brand  # Сохраняем переданную марку внутри конкретного объекта.
        self.color = color  # Сохраняем переданный цвет внутри конкретного объекта.
        self.speed = 0  # Устанавливаем каждой новой машине начальную скорость ноль.

my_car = Car("Toyota", "красный")  # Создаём красную Toyota.
your_car = Car("BMW", "синий")  # Создаём синий BMW.

print(my_car.brand, my_car.color, my_car.speed)  # Показываем данные первой машины.
print(your_car.brand, your_car.color, your_car.speed)  # Показываем данные второй машины.
````

## 3.1. Как Python передаёт `self`

Когда мы пишем:

```python
my_car.show_info()
```

Python приблизительно преобразует этот вызов в:

```python
Car.show_info(my_car)
```

То есть объект `my_car` автоматически передаётся в параметр `self`.

Поэтому при объявлении метода `self` нужно писать, а при обычном вызове — не нужно.

## 3.2. Методы — действия объекта

**Метод** — функция, объявленная внутри класса.

Метод обычно работает с состоянием объекта через `self`.

Для машины создадим методы:

- `accelerate()` — увеличить скорость;
- `brake()` — уменьшить скорость;
- `show_info()` — вернуть текст с информацией.

````python
class Car:  # Создаём класс машины с данными и действиями.
    def __init__(self, brand: str, color: str) -> None:  # Настраиваем новую машину.
        self.brand = brand  # Сохраняем марку машины.
        self.color = color  # Сохраняем цвет машины.
        self.speed = 0  # Устанавливаем начальную скорость.

    def accelerate(self, amount: int) -> None:  # Объявляем метод увеличения скорости.
        self.speed += amount  # Увеличиваем скорость именно текущей машины.

    def brake(self, amount: int) -> None:  # Объявляем метод уменьшения скорости.
        self.speed = max(0, self.speed - amount)  # Уменьшаем скорость, но не допускаем значение ниже нуля.

    def show_info(self) -> str:  # Объявляем метод получения понятного описания машины.
        return f"{self.color} {self.brand}, скорость: {self.speed} км/ч"  # Возвращаем строку с состоянием машины.

my_car = Car("Toyota", "красный")  # Создаём конкретную машину.
print(my_car.show_info())  # Показываем начальное состояние машины.
my_car.accelerate(30)  # Увеличиваем скорость на 30 километров в час.
print(my_car.show_info())  # Показываем изменившееся состояние.
my_car.brake(10)  # Уменьшаем скорость на 10 километров в час.
print(my_car.show_info())  # Показываем итоговое состояние.
````

### Состояние объекта может меняться

После `my_car.accelerate(30)` объект остаётся тем же, но его атрибут `speed` меняется.

Это важная идея ООП:

- объект существует во времени;
- он хранит текущее состояние;
- методы могут безопасно менять это состояние.

Контрольный вопрос: почему метод `accelerate()` не получает параметр `car`? Потому что текущая машина уже доступна через `self`.

# 4. Атрибуты экземпляра и атрибуты класса

В Python есть два распространённых уровня хранения данных.

## Атрибут экземпляра

Создаётся через `self` и принадлежит конкретному объекту.

Примеры:

- `self.color`;
- `self.speed`.

У разных машин эти значения могут отличаться.

## Атрибут класса

Объявляется непосредственно внутри класса и является общим значением для всех объектов.

Например, у всех машин можно задать общее количество колёс: `wheels = 4`.

````python
class Car:  # Создаём класс для сравнения двух видов атрибутов.
    wheels = 4  # Создаём общий атрибут класса: обычная машина имеет четыре колеса.

    def __init__(self, brand: str, color: str) -> None:  # Настраиваем конкретную машину.
        self.brand = brand  # Создаём атрибут экземпляра brand.
        self.color = color  # Создаём атрибут экземпляра color.
        self.speed = 0  # Создаём атрибут экземпляра speed.

car_1 = Car("Toyota", "красный")  # Создаём первую машину.
car_2 = Car("BMW", "синий")  # Создаём вторую машину.

print(car_1.wheels)  # Получаем общий атрибут класса через первый объект.
print(car_2.wheels)  # Получаем тот же общий атрибут через второй объект.
print(Car.wheels)  # Получаем общий атрибут непосредственно через класс.
print(car_1.color)  # Получаем индивидуальный цвет первой машины.
print(car_2.color)  # Получаем индивидуальный цвет второй машины.
````

## 4.1. Как Python ищет атрибут

При обращении `car_1.wheels` Python сначала проверяет сам объект `car_1`.

Если такого атрибута в объекте нет, Python ищет его в классе `Car`.

Это позволяет объектам использовать общие значения класса.

Не следует без необходимости создавать через объект атрибут с тем же именем, что и атрибут класса. Это называется **затенением** и может запутать начинающего разработчика.

````python
print(car_1.wheels)  # Пока значение wheels найдено в классе Car.
car_1.wheels = 3  # Создаём у car_1 собственный атрибут wheels и затеняем значение класса.
print(car_1.wheels)  # Теперь первый объект показывает собственное значение три.
print(car_2.wheels)  # Второй объект продолжает использовать значение класса четыре.
print(Car.wheels)  # Сам класс по-прежнему хранит значение четыре.
````

## 4.2. Опасность изменяемых атрибутов класса

Список, словарь или множество, размещённые на уровне класса, будут общими для всех объектов.

Это часто становится неожиданной ошибкой.

````python
class BadGarage:  # Создаём пример класса с ошибочной общей коллекцией.
    cars = []  # Создаём один общий список для всех гаражей.

    def add_car(self, car_name: str) -> None:  # Объявляем метод добавления машины.
        self.cars.append(car_name)  # Добавляем машину в общий список класса.

garage_1 = BadGarage()  # Создаём первый гараж.
garage_2 = BadGarage()  # Создаём второй гараж.
garage_1.add_car("Toyota")  # Добавляем Toyota через первый гараж.
print(garage_1.cars)  # Первый гараж показывает Toyota.
print(garage_2.cars)  # Второй гараж неожиданно тоже показывает Toyota.
````

````python
class Garage:  # Создаём исправленный класс гаража.
    def __init__(self) -> None:  # Настраиваем каждый новый гараж.
        self.cars = []  # Создаём отдельный список машин для каждого объекта.

    def add_car(self, car_name: str) -> None:  # Объявляем метод добавления машины.
        self.cars.append(car_name)  # Добавляем машину в список конкретного гаража.

garage_1 = Garage()  # Создаём первый независимый гараж.
garage_2 = Garage()  # Создаём второй независимый гараж.
garage_1.add_car("Toyota")  # Добавляем Toyota только в первый гараж.
print(garage_1.cars)  # Первый гараж содержит Toyota.
print(garage_2.cars)  # Второй гараж остаётся пустым.
````

# 5. Инкапсуляция

**Инкапсуляция** — организация объекта таким образом, чтобы он сам контролировал корректность своего состояния.

Простая аналогия: водитель управляет машиной педалями и рулём. Он не меняет напрямую значения внутри блока управления двигателем.

В коде это означает:

- не разрешать любому внешнему коду бесконтрольно менять важные данные;
- предоставлять понятные методы для изменения состояния;
- проверять новые значения;
- скрывать внутренние детали, которые могут измениться.

Например, банковский счёт не должен позволять просто выполнить:

```python
account.balance = -1_000_000
```

Вместо этого объект предоставляет методы `deposit()` и `withdraw()`.

## 5.1. Уровни доступа в Python

В Python нет абсолютной приватности, как в некоторых других языках. Используются соглашения об именовании.

| Запись | Смысл |
|---|---|
| `name` | публичный атрибут, его можно использовать снаружи |
| `_name` | внутренний атрибут, снаружи лучше не использовать |
| `__name` | атрибут с механизмом name mangling, случайный доступ затруднён |

Подчёркивание — сигнал разработчику, а не система безопасности.

````python
class BankAccount:  # Создаём класс банковского счёта с контролем баланса.
    def __init__(self, owner: str, initial_balance: float = 0.0) -> None:  # Настраиваем новый счёт.
        self.owner = owner  # Сохраняем публичное имя владельца.
        self.__balance = initial_balance  # Сохраняем баланс во внутреннем атрибуте с двойным подчёркиванием.

    def deposit(self, amount: float) -> None:  # Объявляем безопасный метод пополнения.
        if amount <= 0:  # Проверяем, что сумма пополнения положительная.
            raise ValueError("Сумма пополнения должна быть больше нуля")  # Сообщаем об ошибочном значении.
        self.__balance += amount  # Увеличиваем внутренний баланс после проверки.

    def withdraw(self, amount: float) -> None:  # Объявляем безопасный метод снятия денег.
        if amount <= 0:  # Проверяем, что сумма снятия положительная.
            raise ValueError("Сумма снятия должна быть больше нуля")  # Сообщаем об ошибочном значении.
        if amount > self.__balance:  # Проверяем, достаточно ли денег на счёте.
            raise ValueError("Недостаточно средств")  # Запрещаем уходить в отрицательный баланс.
        self.__balance -= amount  # Уменьшаем баланс после всех проверок.

    def get_balance(self) -> float:  # Объявляем метод безопасного чтения баланса.
        return self.__balance  # Возвращаем текущее внутреннее значение.

account = BankAccount("Анна", 1000.0)  # Создаём счёт Анны с начальным балансом 1000.
account.deposit(500.0)  # Пополняем счёт на 500.
account.withdraw(300.0)  # Снимаем со счёта 300.
print(account.get_balance())  # Показываем итоговый баланс 1200.
````

### Что именно защищает инкапсуляция

Инкапсуляция защищает не от «хакера», а прежде всего от неправильного использования собственного кода.

Класс `BankAccount` поддерживает **инвариант**: баланс не должен стать отрицательным.

Инвариант — правило, которое должно оставаться истинным для корректного объекта.

Другие примеры инвариантов:

- скорость машины не меньше нуля;
- цена товара не отрицательна;
- дата окончания не раньше даты начала;
- количество товара на складе не отрицательно.

## 5.2. Свойство `property`

Иногда удобно обращаться к значению как к обычному атрибуту, но при этом выполнять проверку.

Для этого используется `property`.

Снаружи код выглядит просто:

```python
car.speed = 50
```

Но внутри срабатывает setter, который может проверить значение.

````python
class SafeCar:  # Создаём машину с контролируемой скоростью.
    def __init__(self, brand: str) -> None:  # Настраиваем новую машину.
        self.brand = brand  # Сохраняем публичную марку.
        self._speed = 0  # Сохраняем внутреннюю скорость в атрибуте с одним подчёркиванием.

    @property  # Превращаем следующий метод в управляемое свойство для чтения.
    def speed(self) -> int:  # Объявляем getter свойства speed.
        return self._speed  # Возвращаем внутреннюю скорость.

    @speed.setter  # Объявляем setter для изменения свойства speed.
    def speed(self, value: int) -> None:  # Получаем новое значение скорости.
        if value < 0:  # Проверяем запрет отрицательной скорости.
            raise ValueError("Скорость не может быть отрицательной")  # Сообщаем об ошибке.
        if value > 250:  # Проверяем учебное ограничение максимальной скорости.
            raise ValueError("Скорость не может быть больше 250 км/ч")  # Сообщаем об ошибке.
        self._speed = value  # Сохраняем проверенное значение.

safe_car = SafeCar("Toyota")  # Создаём безопасную машину.
safe_car.speed = 80  # Вызываем setter и устанавливаем допустимую скорость.
print(safe_car.speed)  # Вызываем getter и читаем скорость.
````

````python
try:  # Начинаем блок демонстрации ожидаемой ошибки.
    safe_car.speed = -10  # Пытаемся установить недопустимую отрицательную скорость.
except ValueError as error:  # Перехватываем понятную ошибку значения.
    print(error)  # Выводим сообщение вместо остановки всего ноутбука.
````

# 6. Наследование

**Наследование** — механизм, при котором новый класс получает атрибуты и методы существующего класса.

Существующий класс называют:

- базовым классом;
- родительским классом;
- суперклассом.

Новый класс называют:

- производным классом;
- дочерним классом;
- подклассом.

## Простая идея

Допустим, в программе есть разные транспортные средства.

У каждого транспортного средства есть:

- название;
- текущая скорость;
- метод движения;
- метод остановки.

У автомобиля дополнительно есть количество дверей.

У электромобиля дополнительно есть заряд батареи.

Общее поведение можно вынести в класс `Vehicle`, а специальные детали — в дочерние классы.

## 6.1. Отношение «является»

Наследование подходит, когда дочерний объект **является разновидностью** родительского объекта.

Проверка фразой:

- автомобиль **является транспортным средством** — да;
- электромобиль **является автомобилем** — да;
- двигатель **является автомобилем** — нет.

Если фраза «является» звучит неправильно, вероятно, нужна композиция, а не наследование.

````python
class Vehicle:  # Создаём базовый класс транспортного средства.
    def __init__(self, name: str) -> None:  # Настраиваем общий объект транспорта.
        self.name = name  # Сохраняем название транспорта.
        self.speed = 0  # Устанавливаем общую начальную скорость.

    def move(self, speed: int) -> str:  # Объявляем общий метод начала движения.
        self.speed = speed  # Сохраняем переданную скорость.
        return f"{self.name} движется со скоростью {self.speed} км/ч"  # Возвращаем сообщение о движении.

    def stop(self) -> str:  # Объявляем общий метод остановки.
        self.speed = 0  # Устанавливаем скорость ноль.
        return f"{self.name} остановился"  # Возвращаем сообщение об остановке.

class Car(Vehicle):  # Создаём класс Car и наследуем возможности Vehicle.
    def __init__(self, name: str, doors: int) -> None:  # Настраиваем автомобиль с дополнительным параметром doors.
        super().__init__(name)  # Вызываем родительский __init__ для настройки name и speed.
        self.doors = doors  # Добавляем специфичный для автомобиля атрибут количества дверей.

my_car = Car("Toyota", 4)  # Создаём автомобиль как объект дочернего класса.
print(my_car.move(60))  # Используем унаследованный метод move.
print(my_car.stop())  # Используем унаследованный метод stop.
print(my_car.doors)  # Используем собственный атрибут дочернего класса.
print(isinstance(my_car, Car))  # Проверяем принадлежность к классу Car.
print(isinstance(my_car, Vehicle))  # Проверяем, что автомобиль также считается транспортом.
````

### Что наследует дочерний класс

Класс `Car` получил от `Vehicle`:

- атрибуты, созданные родительским `__init__()`;
- метод `move()`;
- метод `stop()`.

При этом `Car` добавил собственный атрибут `doors`.

Наследование позволяет не копировать одинаковый код в каждый класс транспорта.

Но наследование создаёт связь между классами. Изменение родителя может повлиять на всех потомков, поэтому иерархия должна оставаться понятной.

## 6.2. Зачем нужен `super()`

`super()` позволяет обратиться к следующему классу в цепочке наследования.

Чаще всего используется так:

```python
super().__init__(...)
```

Это означает: «выполни родительскую часть настройки объекта».

Без такого вызова дочерний класс должен вручную повторять настройку родителя. Повторение ведёт к ошибкам и усложняет изменение кода.

## 6.3. Переопределение метода

**Переопределение** — создание в дочернем классе метода с тем же именем, что и у родителя.

Это позволяет сохранить общий смысл операции, но изменить детали поведения.

Например:

- обычный транспорт движется;
- автомобиль едет по дороге;
- лодка плывёт по воде.

Метод у всех называется `move()`, но текст и логика отличаются.

````python
class Boat(Vehicle):  # Создаём лодку как другой вид транспорта.
    def move(self, speed: int) -> str:  # Переопределяем родительский метод move.
        self.speed = speed  # Сохраняем скорость лодки.
        return f"{self.name} плывёт со скоростью {self.speed} км/ч"  # Возвращаем специальное сообщение для лодки.

class Bicycle(Vehicle):  # Создаём велосипед как ещё один вид транспорта.
    def move(self, speed: int) -> str:  # Переопределяем общий метод движения.
        self.speed = speed  # Сохраняем скорость велосипеда.
        return f"{self.name} едет благодаря усилиям велосипедиста: {self.speed} км/ч"  # Возвращаем специальное сообщение.

boat = Boat("Катер")  # Создаём объект лодки.
bicycle = Bicycle("Городской велосипед")  # Создаём объект велосипеда.
print(boat.move(25))  # Вызываем версию move класса Boat.
print(bicycle.move(15))  # Вызываем версию move класса Bicycle.
````

## 6.4. Многоуровневое наследование

Наследование может строиться цепочкой:

```text
Vehicle → Car → ElectricCar
```

`ElectricCar` получает возможности и `Vehicle`, и `Car`.

Слишком глубокие цепочки сложны для чтения. На практике лучше ограничиваться небольшими и понятными иерархиями.

````python
class ElectricCar(Car):  # Создаём электромобиль как специальный автомобиль.
    def __init__(self, name: str, doors: int, battery_level: int) -> None:  # Настраиваем электромобиль.
        super().__init__(name, doors)  # Используем настройку обычного автомобиля.
        self.battery_level = battery_level  # Добавляем уровень заряда батареи.

    def charge(self) -> str:  # Объявляем специальный метод зарядки.
        self.battery_level = 100  # Устанавливаем полный заряд.
        return f"{self.name} заряжен до {self.battery_level}%"  # Возвращаем сообщение о зарядке.

electric_car = ElectricCar("Tesla", 4, 40)  # Создаём электромобиль с зарядом 40 процентов.
print(electric_car.move(70))  # Используем унаследованный от Vehicle метод move.
print(electric_car.doors)  # Используем атрибут, добавленный классом Car.
print(electric_car.charge())  # Используем собственный метод ElectricCar.
````

### Когда наследование полезно

Наследование подходит, если:

- существует естественная и стабильная иерархия;
- дочерний объект действительно может использоваться вместо родительского;
- общий код нужно централизовать;
- дочерние классы уточняют, а не ломают поведение родителя.

Наследование лучше не использовать только ради устранения двух одинаковых строк. Иногда простая функция или композиция понятнее.

# 7. Полиморфизм

**Полиморфизм** буквально означает «много форм».

В программировании это способность разных объектов поддерживать одинаковую операцию, но выполнять её по-разному.

## Главная идея

У нас есть разные объекты:

- автомобиль;
- лодка;
- велосипед.

У каждого есть метод `move()`.

Код, который вызывает `move()`, может не знать точный класс объекта. Он просто просит объект двигаться.

Python сам выбирает нужную реализацию метода в зависимости от реального объекта.

## 7.1. Полиморфизм через наследование

Все объекты ниже являются транспортом, но их метод `move()` формирует разный результат.

Один и тот же цикл работает с разными классами.

````python
vehicles = [  # Создаём общий список разных транспортных объектов.
    Car("Toyota", 4),  # Добавляем автомобиль.
    Boat("Катер"),  # Добавляем лодку.
    Bicycle("Велосипед"),  # Добавляем велосипед.
]  # Завершаем список транспорта.

speeds = [60, 25, 15]  # Создаём соответствующие скорости для объектов.

for vehicle, speed in zip(vehicles, speeds):  # Перебираем пары объект-скорость.
    print(vehicle.move(speed))  # Вызываем один метод, но получаем разное поведение.
````

### Почему это удобно

Без полиморфизма пришлось бы проверять тип каждого объекта:

```python
if isinstance(vehicle, Car):
    ...
elif isinstance(vehicle, Boat):
    ...
```

При добавлении нового транспорта такой код пришлось бы переписывать.

С полиморфизмом новый класс просто реализует `move()`, а существующий цикл продолжает работать.

Это снижает количество условных конструкций и делает систему расширяемой.

## 7.2. Полиморфизм без общего родителя: duck typing

Python поддерживает **duck typing** — «утиный тип».

Известная формулировка:

> Если объект ходит как утка и крякает как утка, его можно использовать как утку.

Для Python часто важнее не точный класс, а наличие нужного метода.

Функция может работать с любым объектом, у которого есть подходящий `make_sound()`.

````python
class Dog:  # Создаём класс собаки без общего родителя.
    def make_sound(self) -> str:  # Объявляем метод звука собаки.
        return "Гав"  # Возвращаем звук собаки.

class Cat:  # Создаём класс кошки без общего родителя.
    def make_sound(self) -> str:  # Объявляем метод звука кошки.
        return "Мяу"  # Возвращаем звук кошки.

class AlarmClock:  # Создаём класс будильника, который вообще не является животным.
    def make_sound(self) -> str:  # Объявляем метод с тем же ожидаемым именем.
        return "Дзынь"  # Возвращаем звук будильника.

def play_sound(source: object) -> str:  # Создаём функцию, принимающую произвольный объект.
    return source.make_sound()  # Вызываем ожидаемый метод без проверки конкретного класса.

print(play_sound(Dog()))  # Передаём собаку и получаем Гав.
print(play_sound(Cat()))  # Передаём кошку и получаем Мяу.
print(play_sound(AlarmClock()))  # Передаём будильник и получаем Дзынь.
````

### Наследование и полиморфизм — не одно и то же

Наследование отвечает на вопрос:

> Как один класс связан с другим и что он получает от родителя?

Полиморфизм отвечает на вопрос:

> Можно ли одинаково обращаться к разным объектам?

Полиморфизм может появиться через наследование, но в Python он также возможен через duck typing без общего родителя.

## 7.3. Полиморфизм встроенных операций

Оператор `+` тоже полиморфен:

- для чисел выполняет сложение;
- для строк выполняет склеивание;
- для списков выполняет объединение.

Операция записывается одинаково, но поведение зависит от типов объектов.

````python
print(2 + 3)  # Складываем два числа и получаем число пять.
print("авто" + "мобиль")  # Склеиваем две строки и получаем слово автомобиль.
print([1, 2] + [3, 4])  # Объединяем два списка в один новый список.
````

# 8. Три вида методов

В классе Python часто встречаются три вида методов.

## 1. Обычный метод экземпляра

- первый параметр — `self`;
- работает с конкретным объектом;
- может читать и менять его состояние.

## 2. Классовый метод

- отмечается `@classmethod`;
- первый параметр — `cls`;
- работает с самим классом;
- часто используется как альтернативный конструктор.

## 3. Статический метод

- отмечается `@staticmethod`;
- не получает автоматически ни `self`, ни `cls`;
- логически связан с классом, но не зависит от состояния объекта.

````python
class Car:  # Создаём класс, демонстрирующий три вида методов.
    wheels = 4  # Создаём общий атрибут класса.

    def __init__(self, brand: str, speed: int = 0) -> None:  # Настраиваем конкретную машину.
        self.brand = brand  # Сохраняем марку конкретной машины.
        self.speed = speed  # Сохраняем скорость конкретной машины.

    def accelerate(self, amount: int) -> None:  # Создаём обычный метод экземпляра.
        self.speed += amount  # Меняем состояние конкретного объекта через self.

    @classmethod  # Отмечаем следующий метод как классовый.
    def from_string(cls, text: str) -> "Car":  # Создаём объект из строки вида Марка,Скорость.
        brand, speed_text = text.split(",")  # Разделяем входную строку на две части.
        return cls(brand, int(speed_text))  # Создаём объект текущего класса через cls.

    @staticmethod  # Отмечаем следующий метод как статический.
    def km_to_miles(kilometers: float) -> float:  # Объявляем независимое преобразование единиц.
        return kilometers * 0.621371  # Возвращаем приблизительное количество миль.

car = Car.from_string("Toyota,90")  # Создаём машину через альтернативный классовый конструктор.
car.accelerate(10)  # Вызываем обычный метод и увеличиваем скорость объекта.
print(car.brand, car.speed)  # Показываем состояние созданной машины.
print(Car.km_to_miles(100))  # Вызываем статический метод через класс.
````

## 8.1. Как выбрать вид метода

Задайте три вопроса:

1. Метод должен работать с данными конкретного объекта? Используйте обычный метод и `self`.
2. Метод должен создавать объект или менять состояние класса? Используйте `@classmethod` и `cls`.
3. Метод является вспомогательной функцией, связанной с темой класса? Можно использовать `@staticmethod`.

Не нужно превращать каждую обычную функцию в статический метод. Статический метод оправдан, когда его размещение внутри класса делает структуру программы понятнее.

# 9. Магические методы

Магические методы имеют имена с двойными подчёркиваниями: `__имя__`.

Их также называют:

- dunder methods;
- специальными методами.

Они позволяют объектам взаимодействовать со стандартным синтаксисом Python.

Примеры:

- `__init__()` — настройка объекта;
- `__str__()` — человекочитаемый текст;
- `__repr__()` — техническое представление;
- `__len__()` — поддержка `len()`;
- `__eq__()` — поддержка `==`;
- `__add__()` — поддержка `+`;
- `__iter__()` — поддержка перебора;
- `__call__()` — вызов объекта как функции.

````python
class Car:  # Создаём класс с несколькими магическими методами.
    def __init__(self, brand: str, color: str, mileage: int) -> None:  # Настраиваем новую машину.
        self.brand = brand  # Сохраняем марку.
        self.color = color  # Сохраняем цвет.
        self.mileage = mileage  # Сохраняем пробег.

    def __str__(self) -> str:  # Определяем понятный текст для пользователя.
        return f"{self.color} {self.brand}, пробег {self.mileage} км"  # Возвращаем человекочитаемое описание.

    def __repr__(self) -> str:  # Определяем техническое представление объекта.
        return f"Car(brand={self.brand!r}, color={self.color!r}, mileage={self.mileage!r})"  # Возвращаем строку для разработчика.

    def __eq__(self, other: object) -> bool:  # Определяем правила сравнения двух объектов через ==.
        if not isinstance(other, Car):  # Проверяем, является ли второй объект машиной.
            return NotImplemented  # Передаём Python решение о сравнении с другим типом.
        return (self.brand, self.color, self.mileage) == (other.brand, other.color, other.mileage)  # Сравниваем значимые поля.

car_1 = Car("Toyota", "красный", 10000)  # Создаём первую машину.
car_2 = Car("Toyota", "красный", 10000)  # Создаём вторую машину с такими же данными.
print(car_1)  # Функция print автоматически вызывает __str__.
print(repr(car_1))  # Функция repr вызывает __repr__.
print(car_1 == car_2)  # Оператор == вызывает __eq__ и возвращает True.
````

## 9.1. `__len__()` и собственный контейнер

Класс может вести себя как коллекция.

Создадим парковку, для которой работают:

- `len(parking)`;
- цикл `for`;
- проверка `"Toyota" in parking`.

````python
class Parking:  # Создаём класс небольшой коллекции машин.
    def __init__(self) -> None:  # Настраиваем пустую парковку.
        self._cars: list[str] = []  # Создаём внутренний список названий машин.

    def add(self, car_name: str) -> None:  # Объявляем метод добавления машины.
        self._cars.append(car_name)  # Добавляем название во внутренний список.

    def __len__(self) -> int:  # Определяем поведение функции len.
        return len(self._cars)  # Возвращаем количество машин.

    def __iter__(self):  # Определяем возможность перебора объекта в цикле.
        return iter(self._cars)  # Возвращаем итератор внутреннего списка.

    def __contains__(self, car_name: str) -> bool:  # Определяем поведение оператора in.
        return car_name in self._cars  # Проверяем наличие названия во внутреннем списке.

parking = Parking()  # Создаём пустую парковку.
parking.add("Toyota")  # Добавляем первую машину.
parking.add("BMW")  # Добавляем вторую машину.
print(len(parking))  # Получаем количество машин через __len__.
print("Toyota" in parking)  # Проверяем наличие через __contains__.
for car_name in parking:  # Перебираем парковку благодаря __iter__.
    print(car_name)  # Выводим каждое название машины.
````

# 10. Композиция и агрегация

## Композиция

**Композиция** означает, что один объект состоит из других объектов.

Автомобиль не является двигателем. Автомобиль **содержит двигатель**.

Это отношение «имеет»:

- машина имеет двигатель;
- заказ имеет позиции;
- пользователь имеет адрес;
- отчёт имеет таблицы и графики.

Композиция часто безопаснее наследования, потому что зависимости становятся явными.

````python
class Engine:  # Создаём отдельный класс двигателя.
    def __init__(self, power: int) -> None:  # Настраиваем двигатель.
        self.power = power  # Сохраняем мощность двигателя.
        self.is_running = False  # Сохраняем состояние выключенного двигателя.

    def start(self) -> str:  # Объявляем запуск двигателя.
        self.is_running = True  # Меняем состояние двигателя на запущенное.
        return f"Двигатель мощностью {self.power} л.с. запущен"  # Возвращаем сообщение.

class ComposedCar:  # Создаём автомобиль, содержащий объект двигателя.
    def __init__(self, brand: str, engine: Engine) -> None:  # Получаем готовый двигатель при создании машины.
        self.brand = brand  # Сохраняем марку машины.
        self.engine = engine  # Сохраняем переданный объект двигателя внутри машины.

    def start(self) -> str:  # Объявляем запуск автомобиля.
        return f"{self.brand}: {self.engine.start()}"  # Делегируем реальный запуск объекту двигателя.

engine = Engine(150)  # Создаём двигатель как отдельный объект.
car = ComposedCar("Toyota", engine)  # Передаём двигатель внутрь автомобиля.
print(car.start())  # Запускаем автомобиль, который использует двигатель.
````

## 10.1. Агрегация

Агрегация похожа на композицию, но вложенный объект может существовать независимо.

Например, водитель существует отдельно от машины. Машина может получить объект `Driver`, но уничтожение машины не означает уничтожение водителя.

В Python техническая реализация композиции и агрегации часто выглядит одинаково. Разница находится в смысле и жизненном цикле объектов.

## 10.2. Dependency Injection простыми словами

**Внедрение зависимости** означает, что объект получает нужный ему компонент снаружи, а не создаёт его внутри себя.

В предыдущем примере машина получила готовый `Engine` через параметр конструктора.

Преимущества:

- двигатель легко заменить;
- класс машины меньше зависит от конкретной реализации;
- код легче тестировать;
- объект можно собирать из разных компонентов.

# 11. Множественное наследование

**Множественное наследование** означает, что класс наследуется сразу от нескольких родителей.

Синтаксис:

```python
class Child(ParentA, ParentB):
    ...
```

Это мощный, но потенциально сложный механизм.

Для начинающих лучше использовать его осторожно. Наиболее понятный сценарий — небольшие **mixins**.

Mixin — вспомогательный класс, который добавляет одну конкретную возможность.

````python
class PrintableMixin:  # Создаём mixin для печати краткого описания.
    def print_info(self) -> None:  # Объявляем единственную вспомогательную возможность.
        print(self.get_info())  # Вызываем ожидаемый метод основного класса и печатаем результат.

class SavableMixin:  # Создаём mixin для преобразования объекта в словарь.
    def to_dict(self) -> dict:  # Объявляем вспомогательный метод сериализации.
        return self.__dict__.copy()  # Возвращаем копию атрибутов объекта.

class SmartCar(PrintableMixin, SavableMixin):  # Наследуем машину сразу от двух mixin-классов.
    def __init__(self, brand: str, color: str) -> None:  # Настраиваем умную машину.
        self.brand = brand  # Сохраняем марку.
        self.color = color  # Сохраняем цвет.

    def get_info(self) -> str:  # Реализуем метод, который ожидает PrintableMixin.
        return f"{self.color} {self.brand}"  # Возвращаем краткое описание.

smart_car = SmartCar("Tesla", "белый")  # Создаём объект с возможностями двух mixin-классов.
smart_car.print_info()  # Используем метод PrintableMixin.
print(smart_car.to_dict())  # Используем метод SavableMixin.
````

## 11.1. MRO — порядок поиска методов

При нескольких родителях Python должен решить, где искать метод первым.

**MRO — Method Resolution Order** — порядок разрешения методов.

Посмотреть MRO можно через:

```python
ClassName.mro()
```

Python идёт по классам в указанном порядке и выбирает первую подходящую реализацию.

````python
print([class_item.__name__ for class_item in SmartCar.mro()])  # Показываем понятные имена классов в порядке MRO.
````

## 11.2. Ромбовидное наследование

Ромб возникает, когда два класса наследуются от одного общего родителя, а затем новый класс наследуется от обоих.

```text
      A
     / \
    B   C
     \ /
      D
```

MRO и правильное использование `super()` позволяют вызвать общий родительский код один раз.

````python
class Root:  # Создаём общий корневой класс.
    def show_path(self) -> list[str]:  # Объявляем метод, возвращающий начальную часть пути.
        return ["Root"]  # Возвращаем список с именем корневого класса.

class Left(Root):  # Создаём левую ветвь наследования.
    def show_path(self) -> list[str]:  # Переопределяем метод и продолжаем цепочку.
        return ["Left"] + super().show_path()  # Добавляем имя класса и вызываем следующий метод по MRO.

class Right(Root):  # Создаём правую ветвь наследования.
    def show_path(self) -> list[str]:  # Переопределяем тот же метод.
        return ["Right"] + super().show_path()  # Добавляем имя класса и продолжаем MRO.

class Child(Left, Right):  # Создаём нижний класс ромба.
    def show_path(self) -> list[str]:  # Переопределяем метод в дочернем классе.
        return ["Child"] + super().show_path()  # Запускаем кооперативную цепочку super.

child = Child()  # Создаём объект нижнего класса.
print(child.show_path())  # Показываем фактический порядок вызовов.
print([class_item.__name__ for class_item in Child.mro()])  # Показываем MRO класса Child.
````

### Практические правила множественного наследования

- Используйте небольшие mixin-классы с одной ответственностью.
- Не создавайте сложные иерархии без необходимости.
- Используйте `super()`, а не прямые вызовы родителей.
- Проверяйте `mro()`, если поведение кажется неожиданным.
- Предпочитайте композицию, когда объект просто использует другой компонент.

# 12. Абстрактный класс

**Абстрактный класс** задаёт общий контракт для дочерних классов.

Он может содержать:

- готовые обычные методы;
- общие атрибуты;
- абстрактные методы, которые потомки обязаны реализовать.

Абстрактный класс обычно нельзя использовать для создания непосредственного объекта.

Пример: все фигуры должны уметь вычислять площадь, но формула площади у каждой фигуры своя.

````python
from abc import ABC, abstractmethod  # Импортируем средства абстрактных базовых классов.

class Shape(ABC):  # Создаём абстрактный базовый класс фигуры.
    @abstractmethod  # Отмечаем обязательный для потомков метод.
    def area(self) -> float:  # Объявляем общий контракт вычисления площади.
        pass  # Не задаём формулу, потому что она зависит от конкретной фигуры.

    def describe(self) -> str:  # Создаём готовый обычный метод абстрактного класса.
        return f"Площадь фигуры: {self.area():.2f}"  # Используем обязательный метод area.

class Rectangle(Shape):  # Создаём прямоугольник как конкретную фигуру.
    def __init__(self, width: float, height: float) -> None:  # Настраиваем размеры прямоугольника.
        self.width = width  # Сохраняем ширину.
        self.height = height  # Сохраняем высоту.

    def area(self) -> float:  # Реализуем обязательный абстрактный метод.
        return self.width * self.height  # Возвращаем площадь прямоугольника.

rectangle = Rectangle(5, 3)  # Создаём конкретный прямоугольник.
print(rectangle.area())  # Вычисляем площадь.
print(rectangle.describe())  # Используем готовый метод базового класса.
````

````python
try:  # Начинаем безопасную демонстрацию запрета.
    shape = Shape()  # Пытаемся создать объект незавершённого абстрактного класса.
except TypeError as error:  # Перехватываем ожидаемую ошибку типа.
    print(error)  # Показываем объяснение Python.
````

## 12.1. Зачем нужен абстрактный класс

Абстрактный класс полезен, когда нужно гарантировать единый набор методов.

Например, функция может работать со списком фигур и уверенно вызывать `area()`.

Если разработчик создаст новую фигуру и забудет реализовать `area()`, Python не позволит создать её объект.

Это обнаруживает ошибку раньше, чем программа дойдёт до расчётов.

# 13. Интерфейс в Python

В разных языках слово «интерфейс» может означать отдельную конструкцию языка.

В Python интерфейс — это прежде всего **набор ожидаемых методов и свойств**.

Например, интерфейс «звуковой объект» может означать:

```text
объект должен иметь метод make_sound() -> str
```

Реализовать интерфейс в Python можно несколькими способами:

1. договорённостью и duck typing;
2. абстрактным классом `ABC`;
3. структурным протоколом `Protocol`.

## 13.1. Интерфейс через `Protocol`

`Protocol` позволяет описать требуемую структуру без обязательного наследования.

Объект подходит протоколу, если содержит нужные методы.

Это удобно для статической проверки типов и слабой связанности.

````python
from typing import Protocol, runtime_checkable  # Импортируем инструменты структурных протоколов.

@runtime_checkable  # Разрешаем простую проверку isinstance во время выполнения.
class SoundSource(Protocol):  # Описываем интерфейс источника звука.
    def make_sound(self) -> str:  # Требуем наличие метода make_sound.
        ...  # Не задаём реализацию, потому что протокол описывает только форму.

def announce(source: SoundSource) -> str:  # Объявляем функцию, ожидающую объект интерфейса SoundSource.
    return f"Источник издал звук: {source.make_sound()}"  # Используем метод, гарантированный протоколом.

print(announce(Dog()))  # Передаём собаку, хотя она не наследуется от SoundSource.
print(announce(AlarmClock()))  # Передаём будильник с тем же методом.
print(isinstance(Dog(), SoundSource))  # Проверяем структурное соответствие во время выполнения.
````

## 13.2. ABC или Protocol

Используйте `ABC`, когда:

- нужна единая семейная иерархия;
- есть общий готовый код;
- создание неполной реализации нужно запретить во время выполнения.

Используйте `Protocol`, когда:

- важна только форма объекта;
- классы не должны быть связаны наследованием;
- применяется статическая типизация;
- нужен гибкий duck typing с формальным описанием.

Оба инструмента решают похожую задачу контракта, но делают это по-разному.

# 14. `dataclass`

Для классов, которые в основном хранят данные, Python предлагает декоратор `@dataclass`.

Он автоматически создаёт типовые методы:

- `__init__()`;
- `__repr__()`;
- `__eq__()`.

Это уменьшает шаблонный код.

````python
from dataclasses import dataclass, field  # Импортируем dataclass и безопасную фабрику коллекций.

@dataclass  # Просим Python автоматически создать служебные методы класса.
class SimpleCar:  # Создаём простой класс данных автомобиля.
    brand: str  # Объявляем обязательное поле марки.
    color: str  # Объявляем обязательное поле цвета.
    mileage: int = 0  # Объявляем пробег со значением по умолчанию.
    service_history: list[str] = field(default_factory=list)  # Создаём отдельный список истории для каждого объекта.

car_1 = SimpleCar("Toyota", "красный")  # Создаём первый dataclass-объект.
car_2 = SimpleCar("Toyota", "красный")  # Создаём второй объект с такими же данными.
car_1.service_history.append("Замена масла")  # Добавляем запись только первой машине.
print(car_1)  # Используем автоматически созданный __repr__.
print(car_1 == car_2)  # Используем автоматически созданный __eq__.
print(car_2.service_history)  # Проверяем, что список второй машины остался пустым.
````

## 14.1. `frozen=True`

`frozen=True` создаёт объект, поля которого нельзя обычным способом изменить после создания.

Это полезно для объектов-значений:

- координат;
- денежных сумм;
- настроек;
- идентификаторов.

````python
@dataclass(frozen=True)  # Создаём условно неизменяемый класс координаты.
class Point:  # Объявляем точку на плоскости.
    x: float  # Объявляем координату x.
    y: float  # Объявляем координату y.

point = Point(10.0, 20.0)  # Создаём неизменяемую точку.
print(point)  # Выводим автоматически созданное представление.
````

````python
try:  # Начинаем безопасную демонстрацию неизменяемости.
    point.x = 99.0  # Пытаемся изменить замороженное поле.
except Exception as error:  # Перехватываем исключение dataclass без остановки ноутбука.
    print(type(error).__name__, error)  # Показываем тип и текст ошибки.
````

# 15. `__slots__`

По умолчанию объекты Python обычно хранят атрибуты в словаре `__dict__`.

`__slots__` позволяет заранее ограничить набор атрибутов.

Возможные преимущества:

- меньше потребление памяти при очень большом числе объектов;
- запрет случайного создания нового атрибута с опечаткой.

Не следует применять `__slots__` везде. Это инструмент для конкретных задач оптимизации и контроля структуры.

````python
class SlottedCar:  # Создаём класс с фиксированным набором атрибутов.
    __slots__ = ("brand", "speed")  # Разрешаем только brand и speed.

    def __init__(self, brand: str, speed: int = 0) -> None:  # Настраиваем объект.
        self.brand = brand  # Сохраняем разрешённый атрибут brand.
        self.speed = speed  # Сохраняем разрешённый атрибут speed.

slotted_car = SlottedCar("Toyota")  # Создаём объект с ограниченной структурой.
print(slotted_car.brand, slotted_car.speed)  # Читаем разрешённые атрибуты.
````

````python
try:  # Начинаем безопасную демонстрацию ограничения.
    slotted_car.color = "красный"  # Пытаемся создать атрибут, отсутствующий в __slots__.
except AttributeError as error:  # Перехватываем ожидаемую ошибку атрибута.
    print(error)  # Показываем сообщение Python.
````

# 16. Объект как функция: `__call__()`

Если класс реализует `__call__()`, его объект можно вызвать круглыми скобками.

Это удобно для объектов, которые:

- хранят настройку;
- затем многократно применяют её к разным значениям.

Например, ограничитель скорости может хранить максимально допустимую скорость.

````python
class SpeedLimiter:  # Создаём вызываемый объект ограничения скорости.
    def __init__(self, max_speed: int) -> None:  # Сохраняем настройку ограничителя.
        self.max_speed = max_speed  # Запоминаем максимально допустимую скорость.

    def __call__(self, speed: int) -> int:  # Позволяем вызывать объект как функцию.
        return min(speed, self.max_speed)  # Возвращаем скорость не выше ограничения.

limiter = SpeedLimiter(90)  # Создаём ограничитель на 90 километров в час.
print(limiter(70))  # Передаём допустимую скорость и получаем 70.
print(limiter(120))  # Передаём слишком большую скорость и получаем 90.
````

# 17. Контекстный менеджер

Контекстный менеджер используется конструкцией `with`.

Он помогает гарантировать выполнение действий:

- открыть ресурс;
- использовать ресурс;
- корректно закрыть ресурс даже при ошибке.

Стандартный пример — работа с файлом.

Собственный контекстный менеджер реализует:

- `__enter__()` — вход в контекст;
- `__exit__()` — выход из контекста.

````python
class CarWash:  # Создаём учебный контекст автомойки.
    def __enter__(self) -> "CarWash":  # Определяем действие при входе в with.
        print("Мойка открыта")  # Показываем начало работы ресурса.
        return self  # Возвращаем объект для использования внутри блока.

    def wash(self, car_name: str) -> None:  # Объявляем полезное действие внутри контекста.
        print(f"Моем автомобиль {car_name}")  # Имитируем мойку автомобиля.

    def __exit__(self, exc_type, exc_value, traceback) -> bool:  # Определяем действие при выходе из with.
        print("Мойка закрыта")  # Гарантированно показываем закрытие ресурса.
        return False  # Не подавляем возможные исключения внутри блока.

with CarWash() as car_wash:  # Входим в контекст и получаем объект автомойки.
    car_wash.wash("Toyota")  # Используем ресурс внутри безопасного блока.
````

# 18. Пользовательские исключения

Иногда стандартного `ValueError` недостаточно для ясного описания предметной ошибки.

Можно создать собственный класс исключения.

Преимущество: вызывающий код может отличить, например, нехватку топлива от неправильного типа данных.

````python
class NotEnoughFuelError(Exception):  # Создаём предметное исключение нехватки топлива.
    pass  # Не добавляем дополнительную логику в учебном примере.

class FuelCar:  # Создаём автомобиль с запасом топлива.
    def __init__(self, brand: str, fuel: float) -> None:  # Настраиваем машину.
        self.brand = brand  # Сохраняем марку.
        self.fuel = fuel  # Сохраняем количество топлива.

    def drive(self, distance: float) -> str:  # Объявляем поездку на заданное расстояние.
        required_fuel = distance / 10  # Используем простое правило: один литр на десять километров.
        if required_fuel > self.fuel:  # Проверяем достаточность топлива.
            raise NotEnoughFuelError("Недостаточно топлива для поездки")  # Выбрасываем предметное исключение.
        self.fuel -= required_fuel  # Уменьшаем топливо после успешной проверки.
        return f"{self.brand} проехал {distance} км"  # Возвращаем сообщение об успешной поездке.

fuel_car = FuelCar("Toyota", 5.0)  # Создаём машину с пятью литрами топлива.
try:  # Начинаем безопасную попытку дальней поездки.
    print(fuel_car.drive(100))  # Пытаемся проехать сто километров, требующих десять литров.
except NotEnoughFuelError as error:  # Перехватываем именно предметную ошибку.
    print(error)  # Показываем понятное сообщение пользователю.
````

# 19. Копирование объектов

Для копирования используются:

- `copy.copy()` — поверхностная копия;
- `copy.deepcopy()` — глубокая копия.

Разница особенно заметна, если объект содержит вложенные списки или другие изменяемые объекты.

````python
import copy  # Импортируем стандартный модуль копирования.

@dataclass  # Создаём простой класс автомобиля с историей ремонта.
class RepairCar:  # Объявляем класс данных.
    brand: str  # Сохраняем марку.
    repairs: list[str] = field(default_factory=list)  # Сохраняем отдельный список ремонтов.

original = RepairCar("Toyota", ["Замена масла"])  # Создаём исходный объект с вложенным списком.
shallow = copy.copy(original)  # Создаём поверхностную копию, разделяющую вложенный список.
deep = copy.deepcopy(original)  # Создаём глубокую копию с отдельным вложенным списком.
original.repairs.append("Замена фильтра")  # Изменяем список исходного объекта.
print(original.repairs)  # Показываем список исходного объекта.
print(shallow.repairs)  # Поверхностная копия видит то же изменение.
print(deep.repairs)  # Глубокая копия сохраняет независимый список.
````

# 20. SOLID простыми словами

SOLID — набор пяти принципов проектирования объектного кода.

Для начинающего важно не заучить английские формулировки, а понять направление.

## S — Single Responsibility

У класса должна быть одна основная ответственность.

`Car` управляет состоянием машины. Сохранение в базу данных и отправку email лучше передать другим объектам.

## O — Open/Closed

Систему желательно расширять новыми классами без постоянного изменения старого проверенного кода.

Полиморфизм транспорта — пример: добавляем `Train`, а общий цикл `vehicle.move()` не меняется.

## L — Liskov Substitution

Дочерний объект должен корректно заменять родительский.

Если функция ожидает `Vehicle`, объект `Car` не должен неожиданно ломать смысл метода `move()`.

## I — Interface Segregation

Лучше несколько маленьких понятных интерфейсов, чем один огромный интерфейс с ненужными методами.

## D — Dependency Inversion

Высокоуровневый код должен зависеть от абстракций, а не от жёстко созданных деталей.

Передача готового `Engine` в `ComposedCar` — простая форма внедрения зависимости.

# 21. Когда наследование использовать не стоит

Наследование не должно быть автоматическим выбором.

Предпочтите композицию, если:

- объект **имеет** другой объект, а не **является** им;
- компонент нужно легко заменить;
- иерархия становится глубокой;
- дочерний класс вынужден отключать методы родителя;
- связь существует только ради повторного использования пары строк.

Пример:

- машина имеет двигатель — композиция;
- машина является транспортом — наследование.

Хорошая модель должна быть понятной человеку, который видит код впервые.

# 22. Итоговый мини-проект: автопарк

Соберём небольшую систему управления автопарком.

Она покажет сразу несколько идей:

- абстрактный класс `FleetVehicle`;
- наследование `CityCar` и `DeliveryVan`;
- полиморфный метод `calculate_trip_cost()`;
- инкапсуляцию пробега;
- композицию автопарка из транспортных объектов;
- магические методы `__len__()` и `__iter__()`.

Задача упрощена специально, чтобы внимание оставалось на ООП, а не на сложной математике.

````python
from abc import ABC, abstractmethod  # Импортируем абстрактные классы для общего контракта транспорта.

class FleetVehicle(ABC):  # Создаём абстрактный класс единицы автопарка.
    def __init__(self, name: str) -> None:  # Настраиваем общую часть транспорта.
        self.name = name  # Сохраняем понятное название.
        self._mileage = 0.0  # Сохраняем внутренний пробег.

    @property  # Создаём управляемое свойство для чтения пробега.
    def mileage(self) -> float:  # Объявляем getter пробега.
        return self._mileage  # Возвращаем внутренний пробег.

    def complete_trip(self, distance: float) -> float:  # Объявляем общий алгоритм завершения поездки.
        if distance <= 0:  # Проверяем корректность расстояния.
            raise ValueError("Расстояние должно быть больше нуля")  # Запрещаем некорректную поездку.
        cost = self.calculate_trip_cost(distance)  # Делегируем расчёт стоимости конкретному подклассу.
        self._mileage += distance  # Увеличиваем пробег после успешного расчёта.
        return cost  # Возвращаем рассчитанную стоимость.

    @abstractmethod  # Отмечаем обязательный полиморфный метод.
    def calculate_trip_cost(self, distance: float) -> float:  # Задаём контракт расчёта стоимости.
        pass  # Оставляем формулу конкретным видам транспорта.

class CityCar(FleetVehicle):  # Создаём городской автомобиль.
    def calculate_trip_cost(self, distance: float) -> float:  # Реализуем стоимость поездки автомобиля.
        return distance * 8.0  # Используем учебный тариф восемь рублей за километр.

class DeliveryVan(FleetVehicle):  # Создаём грузовой фургон.
    def calculate_trip_cost(self, distance: float) -> float:  # Реализуем другую стоимость поездки.
        return distance * 12.0 + 200.0  # Добавляем более высокий тариф и фиксированную стоимость загрузки.

class Fleet:  # Создаём объект автопарка как композицию транспортных средств.
    def __init__(self) -> None:  # Настраиваем пустой автопарк.
        self._vehicles: list[FleetVehicle] = []  # Создаём внутренний список транспорта.

    def add_vehicle(self, vehicle: FleetVehicle) -> None:  # Объявляем добавление транспорта.
        self._vehicles.append(vehicle)  # Добавляем объект, поддерживающий общий контракт.

    def __len__(self) -> int:  # Поддерживаем функцию len.
        return len(self._vehicles)  # Возвращаем количество объектов.

    def __iter__(self):  # Поддерживаем перебор автопарка.
        return iter(self._vehicles)  # Возвращаем итератор внутреннего списка.

fleet = Fleet()  # Создаём пустой автопарк.
fleet.add_vehicle(CityCar("Легковой автомобиль №1"))  # Добавляем городской автомобиль.
fleet.add_vehicle(DeliveryVan("Фургон №1"))  # Добавляем фургон.
print(f"Количество машин: {len(fleet)}")  # Показываем размер автопарка.
for vehicle in fleet:  # Перебираем разные объекты через общий интерфейс.
    trip_cost = vehicle.complete_trip(10.0)  # Выполняем одинаковую поездку, но с разной формулой стоимости.
    print(vehicle.name, trip_cost, vehicle.mileage)  # Показываем название, стоимость и обновлённый пробег.
````

## 22.1. Где в проекте находятся принципы ООП

### Инкапсуляция

Пробег хранится во внутреннем атрибуте `_mileage`. Изменение происходит через `complete_trip()` после проверки расстояния.

### Наследование

`CityCar` и `DeliveryVan` наследуются от `FleetVehicle`.

### Полиморфизм

Один вызов `calculate_trip_cost(distance)` использует разные формулы в разных подклассах.

### Абстракция

`FleetVehicle` задаёт общий смысл транспорта автопарка, не описывая конкретную формулу стоимости.

### Композиция

`Fleet` содержит список объектов транспорта.

## Решение 3–4. Безопасный счётчик

````python
class Counter:  # Создаём класс счётчика.
    def __init__(self, initial_value: int = 0) -> None:  # Получаем начальное значение.
        if initial_value < 0:  # Проверяем инвариант неотрицательного значения.
            raise ValueError("Начальное значение не может быть отрицательным")  # Сообщаем об ошибке.
        self._value = initial_value  # Сохраняем проверенное значение.

    @property  # Создаём свойство для безопасного чтения.
    def value(self) -> int:  # Объявляем getter значения.
        return self._value  # Возвращаем текущее значение.

    def increment(self) -> None:  # Объявляем увеличение счётчика.
        self._value += 1  # Увеличиваем внутреннее значение на один.

    def decrement(self) -> None:  # Объявляем уменьшение счётчика.
        if self._value == 0:  # Проверяем достижение нижней границы.
            return  # Ничего не меняем, если значение уже равно нулю.
        self._value -= 1  # Уменьшаем внутреннее значение на один.

counter = Counter()  # Создаём счётчик с нулевым значением.
counter.increment()  # Увеличиваем значение до одного.
counter.decrement()  # Уменьшаем значение до нуля.
counter.decrement()  # Повторная попытка не делает значение отрицательным.
print(counter.value)  # Показываем итоговый ноль.
````

## Решение 7–8. Классовый и статический методы

````python
class ParsedStudent:  # Создаём класс студента с дополнительными конструкторами.
    def __init__(self, name: str, age: int) -> None:  # Настраиваем объект.
        self.name = name  # Сохраняем имя.
        self.age = age  # Сохраняем возраст.

    @classmethod  # Объявляем альтернативный конструктор.
    def from_string(cls, text: str) -> "ParsedStudent":  # Получаем строку вида Имя,Возраст.
        name, age_text = text.split(",")  # Разделяем строку.
        return cls(name, int(age_text))  # Создаём и возвращаем объект.

    @staticmethod  # Объявляем независимую проверку возраста.
    def is_adult(age: int) -> bool:  # Получаем возраст для проверки.
        return age >= 18  # Возвращаем True для совершеннолетнего возраста.

parsed_student = ParsedStudent.from_string("Анна,19")  # Создаём объект из строки.
print(parsed_student.name, parsed_student.age)  # Показываем разобранные данные.
print(ParsedStudent.is_adult(parsed_student.age))  # Проверяем совершеннолетие статическим методом.
````

## Решение 9. Абстрактный способ оплаты

````python
class PaymentMethod(ABC):  # Создаём абстрактный контракт способа оплаты.
    @abstractmethod  # Отмечаем обязательный метод.
    def pay(self, amount: float) -> str:  # Задаём общую сигнатуру оплаты.
        pass  # Оставляем реализацию конкретному способу.

class CashPayment(PaymentMethod):  # Создаём оплату наличными.
    def pay(self, amount: float) -> str:  # Реализуем обязательный метод.
        return f"Оплачено наличными: {amount:.2f}"  # Возвращаем сообщение.

class CardPayment(PaymentMethod):  # Создаём оплату картой.
    def pay(self, amount: float) -> str:  # Реализуем тот же контракт иначе.
        return f"Оплачено картой: {amount:.2f}"  # Возвращаем другое сообщение.

payments = [CashPayment(), CardPayment()]  # Создаём разные способы оплаты.
for payment in payments:  # Перебираем полиморфные объекты.
    print(payment.pay(1500.0))  # Вызываем одинаковый метод оплаты.
````

## Решение 10. Композиция заказа

````python
@dataclass(frozen=True)  # Создаём неизменяемую позицию заказа.
class OrderItem:  # Объявляем класс одной позиции.
    name: str  # Сохраняем название товара.
    price: float  # Сохраняем цену одной единицы.
    quantity: int  # Сохраняем количество.

    def total(self) -> float:  # Объявляем расчёт суммы позиции.
        return self.price * self.quantity  # Умножаем цену на количество.

class Order:  # Создаём заказ как композицию позиций.
    def __init__(self) -> None:  # Настраиваем пустой заказ.
        self._items: list[OrderItem] = []  # Создаём внутренний список объектов OrderItem.

    def add_item(self, item: OrderItem) -> None:  # Объявляем добавление позиции.
        self._items.append(item)  # Добавляем готовый объект позиции.

    def total(self) -> float:  # Объявляем итоговую сумму заказа.
        return sum(item.total() for item in self._items)  # Складываем суммы всех позиций.

order = Order()  # Создаём пустой заказ.
order.add_item(OrderItem("Ноутбук", 80000.0, 1))  # Добавляем ноутбук.
order.add_item(OrderItem("Мышь", 2000.0, 2))  # Добавляем две мыши.
print(order.total())  # Показываем общую стоимость заказа.
````

```python
class Car:
    wheels = 4

    def __init__(self, brand):
        self.brand = brand
        self.speed = 0

    def accelerate(self, amount):
        self.speed += amount

    @classmethod
    def from_string(cls, text):
        return cls(text)

    @staticmethod
    def is_valid_speed(speed):
        return speed >= 0
```

Ключевые связи:

```text
класс → описание объектов
объект → конкретный экземпляр класса
атрибут → данные объекта
метод → действие объекта
self → текущий объект
инкапсуляция → контроль состояния
наследование → отношение «является»
композиция → отношение «имеет»
полиморфизм → одинаковый вызов, разное поведение
ABC → обязательный контракт через наследование
Protocol → структурный контракт без обязательного наследования
```
