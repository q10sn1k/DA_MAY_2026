## 2. Данные в виде словаря

Начнем с очень маленького набора данных о продажах по месяцам.

В Python словарь хранит данные в формате:

```python
ключ: значение
```

В нашем примере:

- ключ `"Месяц"` хранит список месяцев;
- ключ `"Продажи"` хранит список значений продаж.

Такой словарь уже похож на таблицу: каждый ключ можно представить как название столбца.

````python
# Создадим простую таблицу с данными о продажах
sales_data = {
    "Месяц": ["Янв", "Фев", "Мар", "Апр"],
    "Продажи": [10, 14, 12, 18]
}

sales_data
````

### Что мы видим

Сейчас это еще не таблица pandas. Это обычный словарь Python.

Но структура уже табличная:

| Месяц | Продажи |
|---|---:|
| Янв | 10 |
| Фев | 14 |
| Мар | 12 |
| Апр | 18 |

Такой формат часто используют для демонстрационных данных.

````python
print("Месяцы:", sales_data["Месяц"])
print("Продажи:", sales_data["Продажи"])
print("Тип объекта sales_data:", type(sales_data))
````

## 3. Превращаем словарь в DataFrame

`DataFrame` — это основная табличная структура в pandas.

Проще всего воспринимать `DataFrame` как таблицу, похожую на лист Excel:

- есть строки;
- есть столбцы;
- у каждого столбца есть название;
- у каждой строки есть индекс.

````python
sales_df = pd.DataFrame(sales_data)
sales_df
````

### Разбор результата

Мы получили таблицу из двух столбцов:

- `Месяц` — текстовый столбец;
- `Продажи` — числовой столбец.

Левая часть таблицы — это индекс строки. В нашем случае pandas автоматически создал индексы `0`, `1`, `2`, `3`.

````python
print("Столбцы таблицы:", list(sales_df.columns))
print("Количество строк и столбцов:", sales_df.shape)
````

## 4. Выбираем один столбец

Можно сказать проще:

````python
sales_series = sales_df["Продажи"]

print("Тип всей таблицы:", type(sales_df))
print("Тип одного столбца:", type(sales_series))
````

## 5. Базовая статистика по числовому столбцу

Метод `.describe()` показывает краткую статистику по числовым данным.

Для продаж он помогает быстро понять:

- сколько значений есть в столбце;
- какое среднее значение;
- какое минимальное и максимальное значение;
- как распределены значения.

````python
print("Статистика по продажам:")
print(sales_df["Продажи"].describe())
````

### Как читать `.describe()`

Ключевые строки:

- `count` — количество значений;
- `mean` — среднее значение;
- `std` — стандартное отклонение, то есть насколько значения отличаются от среднего;
- `min` — минимум;
- `max` — максимум.

На начальном этапе достаточно уверенно понимать `count`, `mean`, `min` и `max`.

````python
mean_sales = sales_df["Продажи"].mean()
min_sales = sales_df["Продажи"].min()
max_sales = sales_df["Продажи"].max()

print("Средние продажи:", mean_sales)
print("Минимальные продажи:", min_sales)
print("Максимальные продажи:", max_sales)
````

## 6. Первый график: продажи по месяцам

Таблица показывает точные значения, но график помогает быстрее увидеть динамику.

Линейный график удобно использовать, когда значения идут во времени: месяцы, недели, дни, кварталы.

````python
sales_df.plot(
    x="Месяц",
    y="Продажи",
    kind="line",
    marker="o",
    title="Продажи по месяцам"
)

plt.xlabel("Месяц")
plt.ylabel("Продажи")
plt.grid(True)
plt.tight_layout()
plt.show()
````

### Контрольный вопрос

В каком месяце продажи были максимальными?

Подсказка: можно ответить глазами по графику или рассчитать через pandas.

````python
max_row = sales_df.loc[sales_df["Продажи"].idxmax()]
print("Месяц с максимальными продажами:", max_row["Месяц"])
print("Значение продаж:", max_row["Продажи"])
````

Теперь создадим таблицу, где есть не только месяц и продажи, но и категории товаров и города.

Такая структура ближе к реальной аналитике: продажи обычно нужно смотреть не только в целом, но и в разрезах.

Например:

## 8. Просмотр таблицы и выбор данных

Перед анализом полезно быстро посмотреть:

### Выбираем столбец `Продажи`

## 9. Группировка данных через `groupby`

Группировка нужна, когда мы хотим посчитать показатель отдельно по каждой группе.

Например:

- средние продажи по категориям;
- сумма продаж по городам;
- количество заказов по статусам;
- средний чек по менеджерам.

Логика простая:

1. Выбираем столбец, по которому группируем.
2. Выбираем числовой столбец, который считаем.
3. Выбираем функцию: `sum`, `mean`, `count`, `max`, `min`.

````python
mean_sales_by_category = category_df.groupby("Категория")["Продажи"].mean()

print("Средние продажи по категориям:")
print(mean_sales_by_category)
````

### Сумма продаж по городам

Среднее значение отвечает на вопрос «сколько в среднем».  
Сумма отвечает на вопрос «сколько всего».

В бизнес-аналитике сумма продаж часто важнее среднего, если мы смотрим общий объем выручки.

````python
total_sales_by_city = category_df.groupby("Город")["Продажи"].sum()

print("Суммарные продажи по городам:")
print(total_sales_by_city)
````

### Группировка как таблица

Иногда удобнее получить результат не как `Series`, а как обычную таблицу `DataFrame`.

Для этого можно использовать `.reset_index()`.

````python
total_sales_by_city_table = total_sales_by_city.reset_index()
total_sales_by_city_table
````

## 11. Сводная таблица

Сводная таблица помогает смотреть показатель сразу в двух разрезах.

Например, мы хотим увидеть средние продажи:

- по строкам — категории;
- по столбцам — города.

Это похоже на сводную таблицу в Excel.

````python
pivot_source_data = {
    "Категория": ["A", "A", "B", "B", "C", "C"],
    "Город": ["Москва", "СПб", "Москва", "СПб", "Москва", "СПб"],
    "Продажи": [10, 15, 20, 25, 30, 35]
}

pivot_df = pd.DataFrame(pivot_source_data)

print("Исходная таблица:")
print(pivot_df)
````

### Строим сводную таблицу со средними продажами

Параметры `pd.pivot_table()`:

- `values="Продажи"` — какой показатель считаем;
- `index="Категория"` — что будет в строках;
- `columns="Город"` — что будет в столбцах;
- `aggfunc="mean"` — какую функцию применяем.

````python
pivot_mean = pd.pivot_table(
    pivot_df,
    values="Продажи",
    index="Категория",
    columns="Город",
    aggfunc="mean"
)

pivot_mean
````

### Сводная таблица с суммой продаж

Если нужно посмотреть общий объем продаж, используем `aggfunc="sum"`.

````python
pivot_sum = pd.pivot_table(
    pivot_df,
    values="Продажи",
    index="Категория",
    columns="Город",
    aggfunc="sum"
)

pivot_sum
````

### Типичная ошибка в сводных таблицах

Если неправильно указать название столбца, pandas выдаст ошибку.

Например, если написать `"продажи"` с маленькой буквы вместо `"Продажи"`, это будет другое имя столбца.

В pandas названия столбцов чувствительны к регистру.

````python
print("Список точных названий столбцов:")
print(list(pivot_df.columns))
````

## 12. Matplotlib: из чего состоит график

В Matplotlib часто используются два объекта:

- `Figure` — вся область рисунка;
- `Axes` — конкретная область графика с осями X и Y.

Для начинающего уровня можно запомнить так:

- `fig` — лист, на котором рисуем;
- `ax` — сам график на этом листе.

````python
x = [i for i in range(6)]
y1 = [i for i in x]
y2 = [i ** 2 for i in x]

print("x:", x)
print("y1 = x:", y1)
print("y2 = x**2:", y2)
````

### Линейный график через `fig, ax = plt.subplots()`

Этот способ удобен, когда нужно более явно управлять графиком: названием, подписями осей, сеткой, легендой и аннотациями.

````python
fig, ax = plt.subplots()

ax.plot(x, y1, marker="o", label="y = x")
ax.plot(x, y2, marker="s", label="y = x**2")

ax.set_title("Пример линейного графика")
ax.set_xlabel("Ось X")
ax.set_ylabel("Ось Y")
ax.grid(True)
ax.legend()

ax.annotate(
    "выделим точку",
    xy=(2, 4),
    xytext=(3, 5),
    arrowprops=dict(arrowstyle="->")
)

plt.show()
````

## 13. Простой линейный график через `plt.plot()`

`plt.plot()` — короткий способ построить линейный график.

Он хорошо подходит для простых учебных и быстрых аналитических графиков.

````python
plt.plot(x, y1, marker="o", label="y = x")
plt.plot(x, y2, linestyle="--", marker="s", label="y = x**2")

plt.title("Линейная диаграмма")
plt.xlabel("Ось X")
plt.ylabel("Ось Y")
plt.legend()
plt.grid(True)
plt.show()
````

## 14. Столбчатая диаграмма

Столбчатая диаграмма подходит, когда нужно сравнить категории.

Например:

- продажи по городам;
- количество клиентов по сегментам;
- число заявок по статусам;
- выручку по товарным категориям.

````python
categories = ["A", "B", "C", "D", "E"]
values = np.array([5, 3, 7, 8, 6])

plt.bar(categories, values, label="Значения")

plt.title("Столбчатая диаграмма")
plt.xlabel("Категории")
plt.ylabel("Значения")
plt.legend()
plt.grid(axis="y")
plt.show()
````

### Горизонтальная столбчатая диаграмма

Горизонтальная диаграмма удобна, когда названия категорий длинные или когда нужно визуально сравнить значения по строкам.

````python
plt.barh(categories, values, label="Значения")

plt.title("Горизонтальная столбчатая диаграмма")
plt.xlabel("Значения")
plt.ylabel("Категории")
plt.legend()
plt.grid(axis="x")
plt.show()
````

## 15. Итоговая мини-практика: анализ заказов

Теперь объединим изученные действия в маленький аналитический сценарий.

У нас есть данные о заказах:

- номер заказа;
- город;
- категория товара;
- канал продаж;
- сумма заказа.

Нужно быстро понять:

1. Общую выручку.
2. Средний чек.
3. Выручку по городам.
4. Выручку по категориям и городам через сводную таблицу.
5. Как выглядит выручка по городам на графике.

````python
orders_data = {
    "Заказ": [1001, 1002, 1003, 1004, 1005, 1006, 1007, 1008],
    "Город": ["Москва", "СПб", "Москва", "Казань", "СПб", "Москва", "Казань", "СПб"],
    "Категория": ["Ноутбуки", "Смартфоны", "Ноутбуки", "Аксессуары", "Смартфоны", "Аксессуары", "Ноутбуки", "Аксессуары"],
    "Канал": ["Сайт", "Сайт", "Маркетплейс", "Сайт", "Маркетплейс", "Сайт", "Маркетплейс", "Сайт"],
    "Сумма": [75000, 42000, 81000, 6500, 39000, 8200, 69000, 7100]
}

orders_df = pd.DataFrame(orders_data)
orders_df
````

### Считаем общую выручку и средний чек

- Общая выручка — сумма всех заказов.
- Средний чек — средняя сумма одного заказа.

````python
total_revenue = orders_df["Сумма"].sum()
average_order_value = orders_df["Сумма"].mean()

print("Общая выручка:", total_revenue)
print("Средний чек:", round(average_order_value, 2))
````

### Выручка по городам

Группируем таблицу по столбцу `Город` и считаем сумму по столбцу `Сумма`.

````python
revenue_by_city = orders_df.groupby("Город")["Сумма"].sum().reset_index()
revenue_by_city
````

### Сводная таблица: категории × города

Теперь посмотрим, какие категории дают выручку в каждом городе.

````python
revenue_pivot = pd.pivot_table(
    orders_df,
    values="Сумма",
    index="Категория",
    columns="Город",
    aggfunc="sum",
    fill_value=0
)

revenue_pivot
````

### График выручки по городам

Для сравнения городов подойдет столбчатая диаграмма.

````python
plt.bar(revenue_by_city["Город"], revenue_by_city["Сумма"])

plt.title("Выручка по городам")
plt.xlabel("Город")
plt.ylabel("Выручка")
plt.grid(axis="y")
plt.show()
````
