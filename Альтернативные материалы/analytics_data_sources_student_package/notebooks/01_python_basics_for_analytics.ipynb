## 2. Почему иногда дробные числа выглядят странно

В исходном файле был пример:

```python
8.6 - 5.9
```

Может получиться не `2.7`, а число вроде `2.6999999999999993`.

Это не ошибка вашей логики. Компьютер хранит дробные числа в двоичной форме, поэтому некоторые десятичные дроби представляются приблизительно.

Для обычной аналитики это важно помнить при сравнении и округлении.

````python
8.6 - 5.9
````

Если нам нужен красивый вывод для человека, можно использовать `round()` — округление.

````python
round(8.6 - 5.9, 2)
````

## 4. Переменные

**Переменная** — это имя, под которым мы сохраняем значение.

Можно представить переменную как подписанную коробку:

- в коробке лежит значение;
- на коробке написано имя;
- позже мы можем обратиться к этому имени и получить значение.

Пример:

```python
price = 1200
```

Здесь `price` — имя переменной, а `1200` — значение.

````python
price = 1200
quantity = 15

revenue = price * quantity

print(revenue)
````

### Зачем аналитику переменные

Без переменных пришлось бы каждый раз писать числа вручную.  
С переменными код становится понятнее:

```python
revenue = price * quantity
```

Такую строку легче читать: «выручка равна цене, умноженной на количество».

````python
price = 1500
quantity = 7
revenue = price * quantity

print("Выручка:", revenue)
````

### Правила именования переменных

В Python имя переменной:

- может содержать буквы, цифры и знак `_`;
- не может начинаться с цифры;
- не должно совпадать со служебными словами Python;
- чувствительно к регистру: `Name` и `name` — разные переменные.

Хорошие имена переменных:

```python
price
quantity
total_revenue
client_age
```

Плохие или ошибочные имена:

```python
1_name
client-age
False
```

````python
client_name = "Анна"
client_age = 32

print(client_name)
print(client_age)
````

## 5. Типичная ошибка: переменная еще не создана

В исходном ноутбуке была ячейка:

```python
print(operand3)
```

Если переменная `operand3` еще не была создана, Python выдаст ошибку:

```text
NameError: name 'operand3' is not defined
```

Это означает: «Я не знаю, что такое `operand3`».

Чтобы избежать ошибки, сначала создаем переменную, потом обращаемся к ней.

````python
operand3 = "всем привет"
print(operand3)
````

## 9. Практика: простые расчеты

### Задача 1. Увеличить и уменьшить число

Пользователь вводит натуральное число.  
Нужно вывести:

1. число, увеличенное в 2 раза;
2. число, уменьшенное в 2 раза.

**Пример ввода:**

```text
8
```

**Пример вывода:**

```text
16
4.0
```

````python
number = int(input("Введите число: "))

print(number * 2)
print(number / 2)
````

### Задача 2. Площадь квадрата

Пользователь вводит длину стороны квадрата.  
Нужно вывести площадь квадрата.

Формула:

```text
площадь = сторона × сторона
```

````python
side = float(input("Введите сторону квадрата: "))

area = side ** 2

print("Площадь квадрата:", area)
````

### Задача 3. Площадь и периметр прямоугольника

Пользователь вводит две стороны прямоугольника.  
Нужно вывести площадь и периметр.

Формулы:

```text
площадь = длина × ширина
периметр = 2 × (длина + ширина)
```

````python
length = float(input("Введите длину: "))
width = float(input("Введите ширину: "))

area = length * width
perimeter = 2 * (length + width)

print("Площадь:", area)
print("Периметр:", perimeter)
````

### Задача 4. Сколько секунд прошло между двумя моментами времени

Пользователь вводит два времени в формате:

```text
часы
минуты
секунды
часы
минуты
секунды
```

Нужно определить, сколько секунд прошло между этими моментами.

Идея решения:

1. Перевести первый момент времени в секунды.
2. Перевести второй момент времени в секунды.
3. Найти разницу.

````python
h1 = int(input("Часы первого момента: "))
m1 = int(input("Минуты первого момента: "))
s1 = int(input("Секунды первого момента: "))

h2 = int(input("Часы второго момента: "))
m2 = int(input("Минуты второго момента: "))
s2 = int(input("Секунды второго момента: "))

time_1_seconds = h1 * 3600 + m1 * 60 + s1
time_2_seconds = h2 * 3600 + m2 * 60 + s2

difference = time_2_seconds - time_1_seconds

print("Прошло секунд:", difference)
````

## 11. Условия `if`, `elif`, `else`

Условия позволяют программе принимать решения.

Общая логика:

```python
if условие:
    что сделать, если условие истинно
else:
    что сделать, если условие ложно
```

Очень важно соблюдать отступы.  
В Python отступ показывает, какие команды относятся к условию.

````python
a = 10

if a == 10:
    print("Число равно десяти")
````

````python
a = 11

if a == 10:
    print("Число равно десяти")
else:
    print("Число не равно десяти")
````

### Несколько условий: `elif`

`elif` используется, когда нужно проверить несколько вариантов.

````python
order_amount = 7500

if order_amount >= 10000:
    print("Крупный заказ")
elif order_amount >= 5000:
    print("Средний заказ")
else:
    print("Небольшой заказ")
````

### Пример из аналитики: классификация клиента

Допустим, мы хотим классифицировать клиента по сумме покупок:

- от 100000 — VIP-клиент;

## 12. Частые ошибки начинающих

### Ошибка 1. Использовали переменную до создания

```python
print(sales)
```

Если `sales` еще не создана, будет `NameError`.

### Ошибка 2. Забыли преобразовать ввод в число

```python
a = input()
print(a + 10)
```

Так нельзя, потому что `input()` возвращает строку.  
Нужно так:

```python
a = int(input())
print(a + 10)
```

### Ошибка 3. Нарушили отступы в `if`

```python
if a > 10:
print("Больше 10")
```

После `if` нужен отступ:

```python
if a > 10:
    print("Больше 10")
```

### Ошибка 4. Перепутали `=` и `==`

- `=` — присваивание значения;
- `==` — проверка равенства.

## 13. Итоговая мини-практика

### Сценарий

Вы работаете с простыми данными о продаже товара.

Пользователь вводит:

1. цену товара;
2. количество проданных товаров;
3. процент скидки.

Нужно рассчитать:

- выручку без скидки;
- сумму скидки;
- итоговую выручку после скидки;
- категорию продажи:
  - если итоговая выручка от 100000 — `"крупная продажа"`;
  - если от 30000 — `"средняя продажа"`;
  - иначе — `"небольшая продажа"`.

````python
price = float(input("Введите цену товара: "))
quantity = int(input("Введите количество: "))
discount_percent = float(input("Введите скидку в процентах: "))

revenue_before_discount = price * quantity
discount_amount = revenue_before_discount * discount_percent / 100
revenue_after_discount = revenue_before_discount - discount_amount

print("Выручка без скидки:", revenue_before_discount)
print("Сумма скидки:", discount_amount)
print("Итоговая выручка:", revenue_after_discount)

if revenue_after_discount >= 100000:
    print("Категория: крупная продажа")
elif revenue_after_discount >= 30000:
    print("Категория: средняя продажа")
else:
    print("Категория: небольшая продажа")
````
