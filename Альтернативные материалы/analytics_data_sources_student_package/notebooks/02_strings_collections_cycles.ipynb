# 1. Строки: текстовые данные

В аналитике строки используются постоянно. Это могут быть:

- имена клиентов;
- названия товаров;
- города;
- email-адреса;
- статусы заказов;
- категории товаров.

В Python строка обозначается типом `str` и записывается в кавычках.


````python
client_name = "Иван Петров"
city = "Москва"
status = "новый клиент"

print("Клиент:", client_name)
print("Город:", city)
print("Статус:", status)
print("Тип переменной client_name:", type(client_name))
````

## 1.1. Одинарные, двойные и многострочные строки

Строку можно записывать в одинарных или двойных кавычках. Для многострочного текста используются тройные кавычки.

Многострочные строки удобно использовать для описаний, комментариев, шаблонов сообщений и длинных текстовых фрагментов.


````python
short_text = 'Короткая строка'
message = "Заказ успешно обработан"

long_comment = """
Комментарий менеджера:
Клиент просит связаться после 18:00.
Повторный звонок назначен на завтра.
"""

print(short_text)
print(message)
print(long_comment)
````

## 1.2. Сложение строк: конкатенация

Сложение строк называется **конкатенацией**. Оно соединяет строки в одну.

Важно: Python не добавляет пробелы автоматически. Если нужен пробел, его нужно указать явно.


````python
first_name = "Анна"
last_name = "Смирнова"

full_name_without_space = first_name + last_name
full_name_with_space = first_name + " " + last_name

print("Без пробела:", full_name_without_space)
print("С пробелом:", full_name_with_space)
````

Для начинающих чаще удобнее использовать `f-строки`. Они позволяют аккуратно подставлять значения переменных в текст.


````python
order_id = 1025
amount = 3490

text = f"Заказ №{order_id} на сумму {amount} руб. обработан"
print(text)
````

## 1.3. Умножение строки

Строку можно умножить на число. В результате она повторится указанное количество раз.

Такой прием редко применяется в бизнес-аналитике напрямую, но полезен для понимания того, как Python работает со строками.


````python
separator = "-" * 40
print(separator)
print("Отчет по продажам")
print(separator)
````

## 1.4. Индексы и срезы строк

У каждого символа в строке есть номер — **индекс**.

Особенность Python: счет начинается с нуля.

Для строки `Python` индексы такие:

| Символ | P | y | t | h | o | n |
|---|---:|---:|---:|---:|---:|---:|
| Индекс | 0 | 1 | 2 | 3 | 4 | 5 |
| Отрицательный индекс | -6 | -5 | -4 | -3 | -2 | -1 |


````python
s = "Python"

print("Строка:", s)
print("Второй символ:", s[1])
print("Последний символ:", s[-1])
print("Символы с индексами 2 и 3:", s[2:4])
print("Строка наоборот:", s[::-1])
````

### Как читать срез `s[2:4]`

Срез читается так: взять элементы **с индекса 2 до индекса 4, но не включая 4**.

То есть правая граница в Python не включается.


````python
product_code = "A-2026-MSK-001"

print("Исходный код товара:", product_code)
print("Первые 6 символов:", product_code[:6])
print("Последние 3 символа:", product_code[-3:])
print("Город в коде:", product_code[7:10])
````

## 1.5. Полезные методы строк

Методы строк помогают приводить текст к единому виду. Это важно для очистки данных.

Например, в таблице могут встретиться значения `Москва`, `москва`, ` МОСКВА `. Формально для Python это разные строки, хотя для аналитика это один и тот же город.


````python
s = "  simpleStrinG  "

print("Исходная строка:", repr(s))
print("Длина строки:", len(s))
print("Без пробелов по краям:", s.strip())
print("Верхний регистр:", s.upper())
print("Нижний регистр:", s.lower())
print("Заменить t на TT:", s.replace("t", "TT"))
print("Индекс первого символа i:", s.find("i"))
print("Индекс последнего символа i:", s.rfind("i"))
````

## 1.6. Проверки строк

Некоторые методы возвращают `True` или `False`. Это помогает проверять качество данных.


````python
value_1 = "12345"
value_2 = "Москва"
value_3 = "Order123"

print(value_1, "состоит только из цифр?", value_1.isdigit())
print(value_2, "состоит только из букв?", value_2.isalpha())
print(value_3, "состоит только из букв и цифр?", value_3.isalnum())
````

## 1.7. Разделение строки на части

Метод `.split()` разбивает строку на список элементов. Это часто используется при обработке текстовых данных.


````python
raw_categories = "продукты,одежда,техника,книги"
categories = raw_categories.split(",")

print("Исходная строка:", raw_categories)
print("Список категорий:", categories)
print("Первая категория:", categories[0])
````

# 3. Кортежи: неизменяемые коллекции

Кортеж похож на список, но его нельзя изменять после создания. Он записывается в круглых скобках `()`.

Кортеж удобно использовать там, где данные должны оставаться постоянными: например, координаты, фиксированный период, код и название справочника.


````python
period = ("2026-01-01", "2026-12-31")
print("Период отчета:", period)
print("Начало периода:", period[0])
print("Конец периода:", period[1])
print("Тип переменной period:", type(period))
````

## 3.1. Разбор типичной ошибки: попытка изменить кортеж

В исходном файле была ячейка:

```python
my_tuple[0] = 'измененный элемент'
```

Она вызывает ошибку `TypeError`, потому что кортеж неизменяемый. В учебной версии мы покажем эту ошибку безопасно — через `try/except`.


````python
my_tuple = (23, 42, 40, 36)
print("Кортеж до попытки изменения:", my_tuple)

try:
    my_tuple[0] = "измененный элемент"
except TypeError as error:
    print("Возникла ошибка TypeError:", error)
    print("Причина: кортеж нельзя изменять после создания.")

print("Кортеж после попытки изменения:", my_tuple)
````

## 3.2. Когда использовать список, а когда кортеж

| Ситуация | Лучше использовать |
|---|---|
| Значения будут добавляться, удаляться или изменяться | Список `list` |
| Значения должны оставаться фиксированными | Кортеж `tuple` |
| Нужно хранить уникальные значения | Множество `set` |
| Нужно хранить пары «ключ — значение» | Словарь `dict` |

# 4. Цикл `for`: повторение действий для каждого элемента

Цикл `for` позволяет пройти по всем элементам последовательности: строки, списка, кортежа, множества или словаря.

Общая форма:

```python
for элемент in последовательность:
    действие
```

Отступ внутри цикла обязателен. Именно отступ показывает Python, какие команды относятся к циклу.


````python
for amount in [1200, 2500, 1700, 3100]:
    print("Сумма заказа:", amount)
````

## 4.1. Использование цикла для расчетов

Цикл можно использовать, чтобы применить одно и то же правило к каждому элементу.


````python
sales = [1200, 2500, 1700, 3100]

for amount in sales:
    amount_with_tax = amount * 1.20
    print("Сумма с НДС:", amount_with_tax)
````

## 4.2. Накопление результата в цикле

Частая задача аналитика — пройти по списку и накопить итоговое значение.


````python
sales = [1200, 2500, 1700, 3100]
total = 0

for amount in sales:
    total = total + amount

print("Общая выручка:", total)
print("Проверка через sum():", sum(sales))
````

## 4.3. `enumerate()`: когда нужен и номер, и значение

Иногда нужно вывести не только значение, но и его порядковый номер.


````python
statuses = ["новый", "оплачен", "доставлен"]

for number, status in enumerate(statuses, start=1):
    print(f"{number}. Статус заказа: {status}")
````

# 5. Цикл `while`: повторение, пока условие истинно

Цикл `while` выполняется до тех пор, пока условие остается истинным.

Общая форма:

```python
while условие:
    действие
```

Важно следить, чтобы условие когда-нибудь стало ложным. Иначе получится бесконечный цикл.


````python
i = 0

while i < 4:
    print("Текущее значение i:", i)
    i = i + 1
````

## 5.1. `break`: досрочная остановка цикла

Команда `break` прекращает выполнение цикла.

Это удобно, когда мы ищем нужное значение и хотим остановиться сразу после нахождения.


````python
amounts = [900, 1200, 2500, 7000, 1800]

for amount in amounts:
    print("Проверяем заказ:", amount)
    if amount > 5000:
        print("Найден крупный заказ:", amount)
        break
````

## 5.2. Безопасный аналог исходного бесконечного `while True`

В исходном файле использовался цикл `while True` с остановкой через `break`. Это нормальная конструкция, но для начинающих ее нужно показывать аккуратно.


````python
i = 0

while True:
    if i == 4:
        break
    print(i)
    i = i + 1
````

# 6. Множества: только уникальные значения

Множество `set` хранит только уникальные элементы. Повторы автоматически удаляются.

В аналитике множество удобно использовать, когда нужно получить:

- список уникальных городов;
- список уникальных клиентов;
- список уникальных категорий;
- набор уникальных ошибок или статусов.


````python
cities = ["Москва", "Казань", "Москва", "Самара", "Казань", "Екатеринбург"]
unique_cities = set(cities)

print("Все города:", cities)
print("Уникальные города:", unique_cities)
print("Количество уникальных городов:", len(unique_cities))
````

## 6.1. Создание и пополнение множества

Добавить элемент в множество можно методом `.add()`.


````python
unique_statuses = {"новый", "оплачен", "доставлен"}
print("До добавления:", unique_statuses)

unique_statuses.add("отменен")
unique_statuses.add("оплачен")  # повтор не добавится второй раз

print("После добавления:", unique_statuses)
````

# 7. Словари: пары «ключ — значение»

Словарь `dict` хранит данные в формате:

```python
ключ: значение
```

Это удобно для описания объектов. Например, один клиент может иметь имя, город, возраст и сегмент.


````python
client = {
    "name": "Анна Смирнова",
    "city": "Москва",
    "age": 34,
    "segment": "B2C"
}

print(client)
print("Имя клиента:", client["name"])
print("Город клиента:", client["city"])
````

## 7.1. Доступ к значениям словаря

Получить значение можно по ключу.

Если ключ может отсутствовать, безопаснее использовать метод `.get()`.


````python
ages = {
    "Dmitry": 40,
    "Andrew": 34
}

print("Возраст Andrew:", ages["Andrew"])
print("Возраст Dmitry:", ages["Dmitry"])
print("Возраст Maria:", ages.get("Maria", "нет данных"))
````

## 7.2. Изменение и добавление значений в словаре

Словарь изменяемый: можно добавлять новые ключи и обновлять значения.


````python
order = {
    "order_id": 1025,
    "city": "Москва",
    "amount": 3490,
    "status": "новый"
}

print("До изменения:", order)

order["status"] = "оплачен"
order["manager"] = "Ирина"

print("После изменения:", order)
````

## 7.3. Цикл по словарю

Метод `.items()` позволяет пройти по ключам и значениям одновременно.


````python
client = {
    "name": "Анна Смирнова",
    "city": "Москва",
    "age": 34,
    "segment": "B2C"
}

for key, value in client.items():
    print(key, "->", value)
````

Нужно вывести:

```text
    1

  12321

```


````python
n = 5
numbers = []

for i in range(1, n + 1):
    numbers.append(i)
    print(" " * (n - i), end="")
    print(*numbers, sep="", end="")
    print(*numbers[-2::-1], sep="")
````

# 9. Итоговая мини-практика: простая обработка заказов

Теперь объединим строки, списки, множества, словари и циклы.

У нас есть список заказов. Каждый заказ представлен словарем. Нужно:

- вывести все заказы в понятном виде;
- посчитать общую выручку;
- найти уникальные города;
- посчитать количество заказов со статусом `оплачен`;
- собрать суммы продаж по городам.


````python
orders = [
    {"id": 1001, "city": "Москва", "amount": 2500, "status": "оплачен"},
    {"id": 1002, "city": "Казань", "amount": 1800, "status": "новый"},
    {"id": 1003, "city": "Москва", "amount": 4200, "status": "оплачен"},
    {"id": 1004, "city": "Самара", "amount": 900, "status": "отменен"},
    {"id": 1005, "city": "Казань", "amount": 3100, "status": "оплачен"},
]

print("Список заказов:")
for order in orders:
    print(f"Заказ {order['id']}: {order['city']}, {order['amount']} руб., статус: {order['status']}")
````

````python
total_revenue = 0
unique_cities = set()
paid_orders_count = 0
revenue_by_city = {}

for order in orders:
    total_revenue = total_revenue + order["amount"]
    unique_cities.add(order["city"])
    
    if order["status"] == "оплачен":
        paid_orders_count = paid_orders_count + 1
    
    city = order["city"]
    amount = order["amount"]
    
    if city not in revenue_by_city:
        revenue_by_city[city] = 0
    revenue_by_city[city] = revenue_by_city[city] + amount

print("Общая выручка:", total_revenue)
print("Уникальные города:", unique_cities)
print("Количество оплаченных заказов:", paid_orders_count)
print("Выручка по городам:", revenue_by_city)
````
